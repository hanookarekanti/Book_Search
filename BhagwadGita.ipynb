```python
# Step 1: Import libraries
import pandas as pd
import os
import re
import unicodedata
import random
import spacy
from collections import defaultdict
import numpy as np
import torch
import torch.nn.functional as F
from torch_geometric.nn import GCNConv
from torch_geometric.data import Data
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
import pickle
import networkx as nx
import matplotlib.pyplot as plt
from pyvis.network import Network
from google.colab import files

# Step 2: Check if GPU is available
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# Load spaCy model
nlp = spacy.load("en_core_web_sm", disable=["parser", "ner"])

# Step 3: Create data folder in Colab
if not os.path.exists("data"):
    os.makedirs("data")

# Step 4: Upload the Bhagwad_Gita.csv file to Colab
if not os.path.exists("Bhagwad_Gita.csv"):
    print("Please upload Bhagwad_Gita.csv:")
    uploaded = files.upload()

# Step 5: Define entity mappings, person names, locations, etc.
entity_map = {
    'Arjun': 'Arjuna', 'Krishn': 'Krishna', 'Yudhishtira': 'Yudhishthira',
    'Bheema': 'Bhima', 'Nakul': 'Nakula', 'Sahadev': 'Sahadeva',
    'Draupady': 'Draupadi', 'Kuntee': 'Kunti', 'Madhusudana': 'Krishna',
    'Hrishikesha': 'Krishna', 'Janardana': 'Krishna', 'Vaasudeva': 'Krishna',
    'Vasudeva': 'Krishna', 'Govinda': 'Krishna', 'Hari': 'Krishna', 'Keshava': 'Krishna',
    'Parthasarathi': 'Krishna', 'Yogeshwara': 'Krishna', 'Achyuta': 'Krishna',
    'Narayana': 'Krishna', 'Vishnu': 'Krishna', 'Partha': 'Arjuna',
    'Kaunteya': 'Arjuna', 'Dhananjaya': 'Arjuna', 'Gudakesha': 'Arjuna',
    'Bharatarshabha': 'Arjuna', 'Kiriti': 'Arjuna', 'Savyasachi': 'Arjuna',
    'Vijaya': 'Arjuna', 'Pritha': 'Kunti', 'Devavrata': 'Bhishma',
    'Gangaputra': 'Bhishma', 'Pitamaha': 'Bhishma', 'Acharya Drona': 'Drona',
    'Kripacharya': 'Kripa', 'Veda Vyasa': 'Vyasa', 'Krishna Dvaipayana': 'Vyasa',
    'Radheya': 'Karna', 'Vaikartana': 'Karna', 'Suryaputra': 'Karna',
    'Angaraja': 'Karna', 'Ashwathama': 'Asvatthama', 'Ashvatama': 'Asvatthama',
    'Shikhandini': 'Sikhandi', 'Alambusha': 'Alambusha', 'Bahlika': 'Bahlika',
    'Shrutayudha': 'Shrutayudha', 'Sudakshina': 'Sudakshina', 'Sarvaga': 'Sarvaga',
    'Pradyumna': 'Pradyumna', 'Samba': 'Samba', 'Satyavati': 'Satyavati',
    'Amba': 'Amba', 'Ambika': 'Ambika', 'Ambalika': 'Ambalika',
    'Phalguna': 'Arjuna', 'Jishnu': 'Arjuna', 'Kiritin': 'Arjuna', 'Pandava': 'Arjuna',
    'Madhava': 'Krishna', 'Damodara': 'Krishna', 'Yadava': 'Krishna',
    'Dharmaraja': 'Yudhishthira', 'Ajatasatru': 'Yudhishthira',
    'Panchali': 'Draupadi', 'Krishnaa': 'Draupadi', 'Yajnaseni': 'Draupadi',
    'Dhrishtadyumna': 'Dhrishtadyumna', 'Duryodhan': 'Duryodhana', 'Suyodhana': 'Duryodhana',
    'Kururaja': 'Duryodhana', 'Gandhari-putra': 'Duryodhana', 'Dhritarashtraputra': 'Duryodhana',
    'Bheeshma': 'Bhishma', 'Shantanava': 'Bhishma', 'Gandiva-dhari': 'Arjuna',
    'Vrikodara': 'Bhima', 'Bhimsen': 'Bhima', 'Subhadraa': 'Subhadra', 'Yadavi': 'Subhadra',
    'Balaram': 'Balarama', 'Baladeva': 'Balarama', 'Sankarshana': 'Balarama',
    'Gandhariputra': 'Dushasana', 'Dussasana': 'Dushasana', 'Madriputra': 'Nakula',
    'Madreyi-suta': 'Sahadeva', 'Kaurava': 'Duryodhana', 'Shakuni': 'Shakuni',
    'Gandhara-raja': 'Shakuni', 'Abhimanyu': 'Abhimanyu', 'Saubhadra': 'Abhimanyu',
    'Parikshit': 'Parikshit', 'Janamejaya': 'Janamejaya', 'Ghatotkacha': 'Ghatotkacha',
    'Hidimba-suta': 'Ghatotkacha', 'Babhruvahana': 'Babhruvahana', 'Iravan': 'Iravan',
    'Uloopi-suta': 'Iravan', 'Chitrangada-suta': 'Babhruvahana', 'Vrishasena': 'Vrishasena',
    'Sushena': 'Sushena', 'Uttaraa': 'Uttara', 'Virataputri': 'Uttara',
    'Bhanumati': 'Bhanumati', 'Vrushali': 'Vrushali', 'Hidimbi': 'Hidimbi',
    'Uloopi': 'Ulupi', 'Chitrangada': 'Chitrangada', 'Devika': 'Devika',
    'Karenumati': 'Karenumati', 'Vijayaa': 'Vijaya', 'Ganga': 'Ganga',
    'Shantanu': 'Shantanu', 'Dhritarastra': 'Dhritarashtra', 'Pandoo': 'Pandu',
    'Madri': 'Madri', 'Gandhari': 'Gandhari', 'Devaki': 'Devaki',
    'Vasudev': 'Vasudeva', 'Dushala': 'Dushala', 'Jayadratha': 'Jayadratha',
    'Satyaki': 'Satyaki', 'Yuyutsu': 'Yuyutsu', 'Shalya': 'Shalya',
    'Dhrishtadyumna': 'Dhrishtadyumna', 'Draupada': 'Drupada', 'Virata': 'Virata',
    'Somadatta': 'Somadatta', 'Bhurisrava': 'Bhurisrava', 'Laxman': 'Laxman',
    'Chitrasena': 'Chitrasena', 'Chitravarma': 'Chitravarma', 'Yaudheya': 'Yaudheya',
    'Niramitra': 'Niramitra', 'Suhotra': 'Suhotra', 'Shrutakarma': 'Shrutakarma',
    'Sutasoma': 'Sutasoma', 'Shrutasena': 'Shrutasena', 'Shrutakirti': 'Shrutakirti',
    'Prativindhya': 'Prativindhya', 'Vikarna': 'Vikarna', 'Vidura': 'Vidura'
}


person_names = [
    'Dhritarashtra', 'Sanjaya', 'Duryodhana', 'Drona', 'Pandu', 'Arjuna', 'Bhima',
    'Yudhishthira', 'Nakula', 'Sahadeva', 'Krishna', 'Bhishma', 'Karna', 'Kripa',
    'Asvatthama', 'Vikarna', 'Bhurisrava', 'Drupada', 'Virata', 'Satyaki', 'Dhrishtaketu',
    'Chekitana', 'Purujit', 'Kuntibhoja', 'Saibya', 'Yodhamanyu', 'Uttamaujas', 'Abhimanyu',
    'Subhadra', 'Draupadi', 'Kunti', 'Pritha', 'Sikhandi', 'Dhrishtadyumna', 'Jayadratha',
    'Vyasa', 'Brahma', 'Manu', 'Ikshvaku', 'Vivasvan', 'Narada', 'Kapila', 'Bhrigu', 'Usanas',
    'Vaasudeva', 'Vasudeva', 'Madhava', 'Govinda', 'Hari', 'Keshava', 'Janardana', 'Parthasarathi',
    'Yogeshwara', 'Achyuta', 'Narayana', 'Hrishikesha', 'Madhusudana', 'Vishnu', 'Partha', 'Kaunteya',
    'Dhananjaya', 'Gudakesha', 'Bharatarshabha', 'Kiriti', 'Savyasachi', 'Vijaya', 'Dharma Raj',
    'Ajatashatru', 'Vrikodara', 'Suyodhana', 'Dushasana', 'Bhagadatta', 'Devavrata', 'Gangaputra',
    'Pitamaha', 'Acharya Drona', 'Bharadvaja-putra', 'Kripacharya', 'Veda Vyasa', 'Krishna Dvaipayana',
    'Radheya', 'Vaikartana', 'Suryaputra', 'Angaraja', 'Shalya', 'Ghatotkacha', 'Uttara', 'Kritavarma',
    'Markandeya', 'Angira', 'Vashishtha', 'Gautama', 'Kashyapa', 'Chyavana', 'Vishvamitra', 'Shiva',
    'Mahadeva', 'Rudra', 'Indra', 'Yama', 'Mrityu', 'Agni', 'Varuna', 'Vayu', 'Aditya', 'Surya', 'Chandra',
    'Kubera', 'Hanuman', 'Maruti', 'Anjaneya', 'Garuda', 'Vasuki', 'Shesha', 'Devaki', 'Somadatta',
    'Alambusha', 'Bahlika', 'Shrutayudha', 'Sudakshina', 'Sarvaga', 'Pradyumna', 'Samba', 'Satyavati',
    'Amba', 'Ambika', 'Ambalika','Gandhari', 'Madri', 'Shakuni', 'Balarama', 'Parikshit', 'Janamejaya',
    'Hidimbi','Ulupi', 'Chitrangada', 'Devika', 'Karenumati', 'Vijaya', 'Bhanumati', 'Vrushali',
    'Ganga', 'Shantanu', 'Vidura', 'Dushala', 'Yuyutsu', 'Laxman', 'Chitrasena',
    'Chitravarma', 'Yaudheya', 'Niramitra', 'Suhotra', 'Iravan', 'Babhruvahana',
    'Vrishasena', 'Sushena', 'Shrutakarma', 'Sutasoma', 'Shrutasena', 'Shrutakirti',
    'Prativindhya', 'Kshatradeva', 'Kshatradharman', 'Satyajit', 'Kekaya', 'Brihadkshatra',
    'Jarasandha', 'Rukmini', 'Satyabhama', 'Jambavati', 'Rohini', 'Aniruddha',
    'Charudeshna', 'Saurin', 'Sini', 'Vinda', 'Anuvinda', 'Shishupala', 'Rukmi',
    'Kamsa', 'Ekalavya', 'Srinjaya', 'Dhruva', 'Ugrasena', 'Vasudeva (father)', 'Rohini (mother)'
]


locations = [
    'Kurukshetra', 'Dharmakshetra', 'Hastinapura', 'Dwarka', 'Mathura', 'Kasi', 'Kuru',
    'Panchala', 'Matsya', 'Magadha', 'Bharata', 'Varshneya', 'Ganga', 'Yamuna', 'Indraloka',
    'Swarga', 'Brahmaloka', 'Vaikuntha', 'Dharmakshetra Kurukshetra', 'Kaurava', 'Pandava',
    'Bharatavarsha', 'Ayodhya', 'Gandhara', 'Anga', 'Videha', 'Kuru Jangala', 'Indraprastha',
    'Kamyaka', 'Dvaitavana', 'Khandavaprastha', 'Khandava', 'Virata', 'Kaurava camp', 'Pandava camp',
    'Kurus', 'Panchal', 'Bharat', 'Kosal', 'Madra',
    'Chedi', 'Avanti', 'Sindhu', 'Kekaya', 'Trigarta', 'Pragjyotisha', 'Manipura',
    'Vanga', 'Kalinga', 'Kosala', 'Mithila', 'Surasena', 'Bhoja', 'Nishada',
    'Shalva', 'Kamboja', 'Yadava', 'Saraswati', 'Godavari', 'Narmada', 'Sindhu (river)',
    'Kubera’s palace', 'Yamaloka', 'Varunaloka', 'Vayuloka', 'Agniloka', 'Gokula',
    'Vrindavana', 'Barsana', 'Nandagaon', 'Mount Meru', 'Himalaya', 'Mandara',
    'Gandhamadana', 'Badrikashrama', 'Haridwar', 'Prayaga', 'Vishnupada', 'Champa',
    'Pundravardhana', 'Tamralipta', 'Ujjayini', 'Taxila', 'Pushkalavati', 'Rajagriha',
    'Girivraja', 'Pataliputra', 'Kundina', 'Vidarbha', 'Dwaravati', 'Anarta',
    'Saurashtra', 'Satyapura', 'Kanyakubja', 'Ahichhatra', 'Kampilya', 'Ekacakra',
    'Varnavata', 'Upaplavya', 'Matsya kingdom', 'Saindhava', 'Drupada’s palace',
    'Hidimba’s forest', 'Manasa', 'Alaka', 'Amaravati', 'Nagapura', 'Kailasa',
    'Patala', 'Rakshasa loka', 'Yaksha loka', 'Gandharva loka'
]


relation_keywords = [
    'son', 'sons', 'daughter', 'daughters', 'father', 'fathers', 'mother', 'brother',
    'brothers', 'sister', 'sisters', 'cousin', 'cousins', 'grandfather', 'grandfathers',
    'grandsire', 'grandson', 'grandsons', 'maternal uncle', 'maternal uncles',
    'father-in-law', 'brother-in-law', 'husband', 'wife', 'child', 'children',
    'parent', 'parents', 'offspring', 'sibling', 'siblings', 'brethren', 'son of',
    'daughter of', 'born to', 'born of', 'children of', 'child of', 'kinsman',
    'heir', 'descendant', 'descendants', 'progeny', 'ancestors', 'lineage', 'born',
    'parentage', 'forefather', 'progenitor', 'scion', 'issue', 'begotten', 'sire',
    'ancestry', 'heritage', 'clan', 'tribe', 'household', 'line', 'race', 'blood',
    'grandmother', 'grandmothers', 'granddaughter', 'granddaughters', 'uncle', 'uncles',
    'aunt', 'aunts', 'nephew', 'nephews', 'niece', 'nieces', 'mother-in-law',
    'sister-in-law', 'son-in-law', 'daughter-in-law', 'stepbrother', 'stepbrothers',
    'stepsister', 'stepsisters', 'stepfather', 'stepmother', 'relative', 'relatives',
    'kin', 'kindred', 'family', 'house', 'dynasty', 'bloodline', 'stock', 'seed',
    'forebear', 'forebears', 'successor', 'successors', 'inheritor', 'inheritors',
    'offspring of', 'issue of', 'spawn', 'get', 'posterity', 'generation', 'generations',
    'patriarch', 'matriarch', 'elder', 'elders', 'younger', 'youngers', 'consort',
    'spouse', 'betrothed', 'fiancé', 'fiancée', 'ward', 'guardian', 'mentor', 'disciple',
    'teacher', 'pupil', 'master', 'student', 'ally', 'allies', 'friend', 'friends',
    'companion', 'companions', 'comrade', 'comrades', 'foe', 'foes', 'enemy', 'enemies',
    'rival', 'rivals', 'chieftain', 'leader', 'follower', 'followers', 'vassal', 'vassals',
    'lord', 'liege', 'retainer', 'retainers', 'patron', 'client', 'kinswoman', 'foremother',
    'clan leader', 'tribal chief', 'house head', 'lineal', 'collateral', 'agnate', 'cognate'
]

gender_map = {
    'Arjuna': 'male', 'Bhima': 'male', 'Yudhishthira': 'male', 'Nakula': 'male',
    'Sahadeva': 'male', 'Krishna': 'male', 'Duryodhana': 'male', 'Bhishma': 'male',
    'Karna': 'male', 'Drona': 'male', 'Dhritarashtra': 'male', 'Sanjaya': 'male',
    'Abhimanyu': 'male', 'Drupada': 'male', 'Virata': 'male', 'Satyaki': 'male',
    'Pandu': 'male', 'Vyasa': 'male', 'Brahma': 'male', 'Manu': 'male',
    'Ikshvaku': 'male', 'Vivasvan': 'male', 'Narada': 'male', 'Kapila': 'male',
    'Bhrigu': 'male', 'Usanas': 'male', 'Chekitana': 'male', 'Dhrishtaketu': 'male',
    'Yodhamanyu': 'male', 'Kripa': 'male', 'Asvatthama': 'male', 'Bhurisrava': 'male',
    'Somadatta': 'male', 'Kunti': 'female', 'Subhadra': 'female', 'Draupadi': 'female',
    'Devaki': 'female', 'Sikhandi': 'male',
    'Dhrishtadyumna': 'male', 'Jayadratha': 'male', 'Shalya': 'male', 'Ghatotkacha': 'male',
    'Kritavarma': 'male', 'Markandeya': 'male', 'Angira': 'male', 'Vashishtha': 'male',
    'Gautama': 'male', 'Kashyapa': 'male', 'Chyavana': 'male', 'Vishvamitra': 'male',
    'Shiva': 'male', 'Indra': 'male', 'Yama': 'male', 'Mrityu': 'male', 'Agni': 'male',
    'Varuna': 'male', 'Vayu': 'male', 'Aditya': 'male', 'Surya': 'male', 'Chandra': 'male',
    'Kubera': 'male', 'Hanuman': 'male', 'Garuda': 'male', 'Vasuki': 'male', 'Shesha': 'male',
    'Alambusha': 'male', 'Bahlika': 'male', 'Shrutayudha': 'male', 'Sudakshina': 'male',
    'Sarvaga': 'male', 'Pradyumna': 'male', 'Samba': 'male', 'Satyavati': 'female',
    'Amba': 'female', 'Ambika': 'female', 'Ambalika': 'female',
    'Gandhari': 'female', 'Madri': 'female', 'Shakuni': 'male', 'Balarama': 'male',
    'Parikshit': 'male', 'Janamejaya': 'male', 'Hidimbi': 'female', 'Ulupi': 'female',
    'Chitrangada': 'female', 'Devika': 'female', 'Karenumati': 'female', 'Vijaya': 'female',
    'Bhanumati': 'female', 'Vrushali': 'female', 'Ganga': 'female', 'Shantanu': 'male',
    'Vidura': 'male', 'Dushala': 'female', 'Yuyutsu': 'male', 'Laxman': 'male',
    'Chitrasena': 'male', 'Chitravarma': 'male', 'Yaudheya': 'male', 'Niramitra': 'male',
    'Suhotra': 'male', 'Iravan': 'male', 'Babhruvahana': 'male', 'Vrishasena': 'male',
    'Sushena': 'male', 'Uttara': 'female',
    'Shrutakarma': 'male', 'Sutasoma': 'male', 'Shrutasena': 'male', 'Shrutakirti': 'male',
    'Prativindhya': 'male', 'Kshatradeva': 'male', 'Kshatradharman': 'male', 'Satyajit': 'male',
    'Kekaya': 'male', 'Brihadkshatra': 'male', 'Jarasandha': 'male', 'Rukmini': 'female',
    'Satyabhama': 'female', 'Jambavati': 'female', 'Rohini': 'female', 'Aniruddha': 'male',
    'Charudeshna': 'male', 'Saurin': 'male', 'Sini': 'male', 'Vinda': 'male', 'Anuvinda': 'male',
    'Shishupala': 'male', 'Rukmi': 'male', 'Kamsa': 'male', 'Ekalavya': 'male', 'Srinjaya': 'male',
    'Dhruva': 'male', 'Ugrasena': 'male', 'Vasudeva (father)': 'male', 'Rohini (mother)': 'female'
}

# Precompute lowercase sets for faster lookups
person_names_set = set(name.lower() for name in person_names)
locations_set = set(loc.lower() for loc in locations)
relation_keywords_set = set(rel.lower() for rel in relation_keywords)

# Normalize person names and gender map
name_to_canonical = {k.lower(): v.lower() for k, v in entity_map.items()}
normalized_person_names = []
for name in person_names:
    name_lower = name.lower()
    if name_lower in name_to_canonical:
        normalized_person_names.append(name_to_canonical[name_lower])
    else:
        normalized_person_names.append(name_lower)
person_names = list(dict.fromkeys(normalized_person_names))  # Remove duplicates
person_names_set = set(person_names)  # Update the set with normalized names

gender_map = {k.lower(): v for k, v in gender_map.items()}
for k in list(gender_map.keys()):
    if k in name_to_canonical:
        gender_map[name_to_canonical[k]] = gender_map.pop(k)

# Step 6: Define predefined relations (for Code 2)
predefined_relations = [
    # Parent-Child Relations
    ('pandu', 'father', 'arjuna'), ('arjuna', 'son', 'pandu'),
    ('pandu', 'father', 'bhima'), ('bhima', 'son', 'pandu'),
    ('pandu', 'father', 'yudhishthira'), ('yudhishthira', 'son', 'pandu'),
    ('pandu', 'father', 'nakula'), ('nakula', 'son', 'pandu'),
    ('pandu', 'father', 'sahadeva'), ('sahadeva', 'son', 'pandu'),
    ('kunti', 'mother', 'arjuna'), ('arjuna', 'son', 'kunti'),
    ('kunti', 'mother', 'bhima'), ('bhima', 'son', 'kunti'),
    ('kunti', 'mother', 'yudhishthira'), ('yudhishthira', 'son', 'kunti'),
    ('kunti', 'mother', 'karna'), ('karna', 'son', 'kunti'),
    ('madri', 'mother', 'nakula'), ('nakula', 'son', 'madri'),
    ('madri', 'mother', 'sahadeva'), ('sahadeva', 'son', 'madri'),
    ('dhritarashtra', 'father', 'duryodhana'), ('duryodhana', 'son', 'dhritarashtra'),
    ('dhritarashtra', 'father', 'dushasana'), ('dushasana', 'son', 'dhritarashtra'),
    ('dhritarashtra', 'father', 'vikarna'), ('vikarna', 'son', 'dhritarashtra'),
    ('gandhari', 'mother', 'duryodhana'), ('duryodhana', 'son', 'gandhari'),
    ('gandhari', 'mother', 'dushasana'), ('dushasana', 'son', 'gandhari'),
    ('gandhari', 'mother', 'vikarna'), ('vikarna', 'son', 'gandhari'),
    ('drupada', 'father', 'draupadi'), ('draupadi', 'daughter', 'drupada'),
    ('drupada', 'father', 'dhrishtadyumna'), ('dhrishtadyumna', 'son', 'drupada'),
    ('vasudeva', 'father', 'krishna'), ('krishna', 'son', 'vasudeva'),
    ('vasudeva', 'father', 'subhadra'), ('subhadra', 'daughter', 'vasudeva'),
    ('vasudeva', 'father', 'balarama'), ('balarama', 'son', 'vasudeva'),
    ('devaki', 'mother', 'krishna'), ('krishna', 'son', 'devaki'),
    ('devaki', 'mother', 'subhadra'), ('subhadra', 'daughter', 'devaki'),
    ('devaki', 'mother', 'balarama'), ('balarama', 'son', 'devaki'),
    ('somadatta', 'father', 'bhurisrava'), ('bhurisrava', 'son', 'somadatta'),
    ('drona', 'father', 'asvatthama'), ('asvatthama', 'son', 'drona'),
    ('ganga', 'mother', 'bhishma'), ('bhishma', 'son', 'ganga'),
    ('shantanu', 'father', 'bhishma'), ('bhishma', 'son', 'shantanu'),
    ('shantanu', 'father', 'chitrangada'), ('chitrangada', 'son', 'shantanu'),
    ('shantanu', 'father', 'vichitravirya'), ('vichitravirya', 'son', 'shantanu'),
    ('satyavati', 'mother', 'vyasa'), ('vyasa', 'son', 'satyavati'),
    ('parashara', 'father', 'vyasa'), ('vyasa', 'son', 'parashara'),
    ('vyasa', 'father', 'dhritarashtra'), ('dhritarashtra', 'son', 'vyasa'),
    ('vyasa', 'father', 'pandu'), ('pandu', 'son', 'vyasa'),
    ('vyasa', 'father', 'vidura'), ('vidura', 'son', 'vyasa'),
    ('arjuna', 'father', 'abhimanyu'), ('abhimanyu', 'son', 'arjuna'),
    ('subhadra', 'mother', 'abhimanyu'), ('abhimanyu', 'son', 'subhadra'),
    ('bhima', 'father', 'ghatotkacha'), ('ghatotkacha', 'son', 'bhima'),
    ('hidimbi', 'mother', 'ghatotkacha'), ('ghatotkacha', 'son', 'hidimbi'),
    ('yudhishthira', 'father', 'yaudheya'), ('yaudheya', 'son', 'yudhishthira'),
    ('devika', 'mother', 'yaudheya'), ('yaudheya', 'son', 'devika'),
    ('nakula', 'father', 'niramitra'), ('niramitra', 'son', 'nakula'),
    ('karenumati', 'mother', 'niramitra'), ('niramitra', 'son', 'karenumati'),
    ('sahadeva', 'father', 'suhotra'), ('suhotra', 'son', 'sahadeva'),
    ('vijaya', 'mother', 'suhotra'), ('suhotra', 'son', 'vijaya'),
    ('karna', 'father', 'vrishasena'), ('vrishasena', 'son', 'karna'),
    ('karna', 'father', 'sushena'), ('sushena', 'son', 'karna'),
    ('arjuna', 'father', 'iravan'), ('iravan', 'son', 'arjuna'),
    ('ulupi', 'mother', 'iravan'), ('iravan', 'son', 'ulupi'),
    ('arjuna', 'father', 'babhruvahana'), ('babhruvahana', 'son', 'arjuna'),
    ('chitrangada', 'mother', 'babhruvahana'), ('babhruvahana', 'son', 'chitrangada'),
    ('abhimanyu', 'father', 'parikshit'), ('parikshit', 'son', 'abhimanyu'),
    ('uttara', 'mother', 'parikshit'), ('parikshit', 'son', 'uttara'),
    ('duryodhana', 'father', 'laxman'), ('laxman', 'son', 'duryodhana'),
    ('bhanumati', 'mother', 'laxman'), ('laxman', 'son', 'bhanumati'),
    ('dushasana', 'father', 'chitrasena'), ('chitrasena', 'son', 'dushasana'),
    ('vikarna', 'father', 'chitravarma'), ('chitravarma', 'son', 'vikarna'),
    ('vichitravirya', 'father', 'dhritarashtra'), ('dhritarashtra', 'son', 'vichitravirya'),
    ('vichitravirya', 'father', 'pandu'), ('pandu', 'son', 'vichitravirya'),
    ('ambika', 'mother', 'dhritarashtra'), ('dhritarashtra', 'son', 'ambika'),
    ('ambalika', 'mother', 'pandu'), ('pandu', 'son', 'ambalika'),

    # Sibling Relations
    ('arjuna', 'brother', 'bhima'), ('bhima', 'brother', 'arjuna'),
    ('arjuna', 'brother', 'yudhishthira'), ('yudhishthira', 'brother', 'arjuna'),
    ('arjuna', 'brother', 'nakula'), ('nakula', 'brother', 'arjuna'),
    ('arjuna', 'brother', 'sahadeva'), ('sahadeva', 'brother', 'arjuna'),
    ('arjuna', 'brother', 'karna'), ('karna', 'brother', 'arjuna'),
    ('bhima', 'brother', 'yudhishthira'), ('yudhishthira', 'brother', 'bhima'),
    ('bhima', 'brother', 'nakula'), ('nakula', 'brother', 'bhima'),
    ('bhima', 'brother', 'sahadeva'), ('sahadeva', 'brother', 'bhima'),
    ('bhima', 'brother', 'karna'), ('karna', 'brother', 'bhima'),
    ('yudhishthira', 'brother', 'nakula'), ('nakula', 'brother', 'yudhishthira'),
    ('yudhishthira', 'brother', 'sahadeva'), ('sahadeva', 'brother', 'yudhishthira'),
    ('yudhishthira', 'brother', 'karna'), ('karna', 'brother', 'yudhishthira'),
    ('nakula', 'brother', 'sahadeva'), ('sahadeva', 'brother', 'nakula'),
    ('nakula', 'brother', 'karna'), ('karna', 'brother', 'nakula'),
    ('sahadeva', 'brother', 'karna'), ('karna', 'brother', 'sahadeva'),
    ('duryodhana', 'brother', 'dushasana'), ('dushasana', 'brother', 'duryodhana'),
    ('duryodhana', 'brother', 'vikarna'), ('vikarna', 'brother', 'duryodhana'),
    ('duryodhana', 'brother', 'dushala'), ('dushala', 'sister', 'duryodhana'),
    ('dushasana', 'brother', 'vikarna'), ('vikarna', 'brother', 'dushasana'),
    ('dushasana', 'brother', 'dushala'), ('dushala', 'sister', 'dushasana'),
    ('vikarna', 'brother', 'dushala'), ('dushala', 'sister', 'vikarna'),
    ('krishna', 'brother', 'subhadra'), ('subhadra', 'sister', 'krishna'),
    ('krishna', 'brother', 'balarama'), ('balarama', 'brother', 'krishna'),
    ('subhadra', 'sister', 'balarama'), ('balarama', 'brother', 'subhadra'),
    ('draupadi', 'sister', 'dhrishtadyumna'), ('dhrishtadyumna', 'brother', 'draupadi'),
    ('dhritarashtra', 'brother', 'pandu'), ('pandu', 'brother', 'dhritarashtra'),
    ('dhritarashtra', 'brother', 'vidura'), ('vidura', 'brother', 'dhritarashtra'),
    ('pandu', 'brother', 'vidura'), ('vidura', 'brother', 'pandu'),
    ('chitrangada', 'brother', 'vichitravirya'), ('vichitravirya', 'brother', 'chitrangada'),

    # Spouse Relations
    ('arjuna', 'husband', 'subhadra'), ('subhadra', 'wife', 'arjuna'),
    ('arjuna', 'husband', 'draupadi'), ('draupadi', 'wife', 'arjuna'),
    ('arjuna', 'husband', 'ulupi'), ('ulupi', 'wife', 'arjuna'),
    ('arjuna', 'husband', 'chitrangada'), ('chitrangada', 'wife', 'arjuna'),
    ('bhima', 'husband', 'hidimbi'), ('hidimbi', 'wife', 'bhima'),
    ('bhima', 'husband', 'draupadi'), ('draupadi', 'wife', 'bhima'),
    ('yudhishthira', 'husband', 'draupadi'), ('draupadi', 'wife', 'yudhishthira'),
    ('yudhishthira', 'husband', 'devika'), ('devika', 'wife', 'yudhishthira'),
    ('nakula', 'husband', 'draupadi'), ('draupadi', 'wife', 'nakula'),
    ('nakula', 'husband', 'karenumati'), ('karenumati', 'wife', 'nakula'),
    ('sahadeva', 'husband', 'draupadi'), ('draupadi', 'wife', 'sahadeva'),
    ('sahadeva', 'husband', 'vijaya'), ('vijaya', 'wife', 'sahadeva'),
    ('duryodhana', 'husband', 'bhanumati'), ('bhanumati', 'wife', 'duryodhana'),
    ('abhimanyu', 'husband', 'uttara'), ('uttara', 'wife', 'abhimanyu'),
    ('karna', 'husband', 'vrushali'), ('vrushali', 'wife', 'karna'),
    ('shantanu', 'husband', 'ganga'), ('ganga', 'wife', 'shantanu'),
    ('shantanu', 'husband', 'satyavati'), ('satyavati', 'wife', 'shantanu'),
    ('vichitravirya', 'husband', 'ambika'), ('ambika', 'wife', 'vichitravirya'),
    ('vichitravirya', 'husband', 'ambalika'), ('ambalika', 'wife', 'vichitravirya'),
    ('dhritarashtra', 'husband', 'gandhari'), ('gandhari', 'wife', 'dhritarashtra'),
    ('pandu', 'husband', 'kunti'), ('kunti', 'wife', 'pandu'),
    ('pandu', 'husband', 'madri'), ('madri', 'wife', 'pandu'),
    ('vasudeva', 'husband', 'devaki'), ('devaki', 'wife', 'vasudeva'),
    ('dushala', 'wife', 'jayadratha'), ('jayadratha', 'husband', 'dushala'),  # Added from invalid_pairs

    # Grandparent Relations
    ('arjuna', 'grandfather', 'parikshit'), ('parikshit', 'grandson', 'arjuna'),
    ('subhadra', 'grandmother', 'parikshit'), ('parikshit', 'grandson', 'subhadra'),
    ('bhima', 'grandfather', 'sutasoma'), ('sutasoma', 'grandson', 'bhima'),
    ('draupadi', 'grandmother', 'sutasoma'), ('sutasoma', 'grandson', 'draupadi'),
    ('yudhishthira', 'grandfather', 'prativindhya'), ('prativindhya', 'grandson', 'yudhishthira'),
    ('draupadi', 'grandmother', 'prativindhya'), ('prativindhya', 'grandson', 'draupadi'),
    ('nakula', 'grandfather', 'shrutasena'), ('shrutasena', 'grandson', 'nakula'),
    ('draupadi', 'grandmother', 'shrutasena'), ('shrutasena', 'grandson', 'draupadi'),
    ('sahadeva', 'grandfather', 'shrutakirti'), ('shrutakirti', 'grandson', 'sahadeva'),
    ('draupadi', 'grandmother', 'shrutakirti'), ('shrutakirti', 'grandson', 'draupadi'),
    ('arjuna', 'grandfather', 'shrutakarma'), ('shrutakarma', 'grandson', 'arjuna'),
    ('draupadi', 'grandmother', 'shrutakarma'), ('shrutakarma', 'grandson', 'draupadi'),
    ('bhishma', 'great-grandfather', 'duryodhana'), ('duryodhana', 'great-grandson', 'bhishma'),
    ('bhishma', 'great-grandfather', 'arjuna'), ('arjuna', 'great-grandson', 'bhishma'),
    ('bhishma', 'great-grandfather', 'bhima'), ('bhima', 'great-grandson', 'bhishma'),
    ('bhishma', 'great-grandfather', 'yudhishthira'), ('yudhishthira', 'great-grandson', 'bhishma'),
    ('bhishma', 'great-grandfather', 'nakula'), ('nakula', 'great-grandson', 'bhishma'),
    ('bhishma', 'great-grandfather', 'sahadeva'), ('sahadeva', 'great-grandson', 'bhishma'),
    ('shantanu', 'great-great-grandfather', 'dhritarashtra'), ('dhritarashtra', 'great-great-grandson', 'shantanu'),
    ('shantanu', 'great-great-grandfather', 'pandu'), ('pandu', 'great-great-grandson', 'shantanu'),
    ('satyavati', 'great-great-grandmother', 'dhritarashtra'), ('dhritarashtra', 'great-great-grandson', 'satyavati'),
    ('satyavati', 'great-great-grandmother', 'pandu'), ('pandu', 'great-great-grandson', 'satyavati'),
    ('vyasa', 'grandfather', 'arjuna'), ('arjuna', 'grandson', 'vyasa'),
    ('vyasa', 'grandfather', 'duryodhana'), ('duryodhana', 'grandson', 'vyasa'),
    ('vyasa', 'grandfather', 'bhima'), ('bhima', 'grandson', 'vyasa'),
    ('vyasa', 'grandfather', 'yudhishthira'), ('yudhishthira', 'grandson', 'vyasa'),
    ('vyasa', 'grandfather', 'nakula'), ('nakula', 'grandson', 'vyasa'),
    ('vyasa', 'grandfather', 'sahadeva'), ('sahadeva', 'grandson', 'vyasa'),
    ('vasudeva', 'great-grandfather', 'parikshit'), ('parikshit', 'great-grandson', 'vasudeva'),
    ('devaki', 'great-grandmother', 'parikshit'), ('parikshit', 'great-grandson', 'devaki'),

    # Children of Draupadi (with all Pandavas)
    ('draupadi', 'mother', 'prativindhya'), ('prativindhya', 'son', 'draupadi'),
    ('yudhishthira', 'father', 'prativindhya'), ('prativindhya', 'son', 'yudhishthira'),
    ('draupadi', 'mother', 'sutasoma'), ('sutasoma', 'son', 'draupadi'),
    ('bhima', 'father', 'sutasoma'), ('sutasoma', 'son', 'bhima'),
    ('draupadi', 'mother', 'shrutakarma'), ('shrutakarma', 'son', 'draupadi'),
    ('arjuna', 'father', 'shrutakarma'), ('shrutakarma', 'son', 'arjuna'),
    ('draupadi', 'mother', 'shrutasena'), ('shrutasena', 'son', 'draupadi'),
    ('nakula', 'father', 'shrutasena'), ('shrutasena', 'son', 'nakula'),
    ('draupadi', 'mother', 'shrutakirti'), ('shrutakirti', 'son', 'draupadi'),
    ('sahadeva', 'father', 'shrutakirti'), ('shrutakirti', 'son', 'sahadeva'),

    # Cousin Relations
    ('arjuna', 'cousin', 'krishna'), ('krishna', 'cousin', 'arjuna'),
    ('bhima', 'cousin', 'krishna'), ('krishna', 'cousin', 'bhima'),
    ('yudhishthira', 'cousin', 'krishna'), ('krishna', 'cousin', 'yudhishthira'),
    ('nakula', 'cousin', 'krishna'), ('krishna', 'cousin', 'nakula'),
    ('sahadeva', 'cousin', 'krishna'), ('krishna', 'cousin', 'sahadeva'),
    ('arjuna', 'cousin', 'balarama'), ('balarama', 'cousin', 'arjuna'),
    ('bhima', 'cousin', 'balarama'), ('balarama', 'cousin', 'bhima'),
    ('yudhishthira', 'cousin', 'balarama'), ('balarama', 'cousin', 'yudhishthira'),
    ('nakula', 'cousin', 'balarama'), ('balarama', 'cousin', 'nakula'),
    ('sahadeva', 'cousin', 'balarama'), ('balarama', 'cousin', 'sahadeva'),
    ('arjuna', 'cousin', 'duryodhana'), ('duryodhana', 'cousin', 'arjuna'),
    ('bhima', 'cousin', 'duryodhana'), ('duryodhana', 'cousin', 'bhima'),
    ('yudhishthira', 'cousin', 'duryodhana'), ('duryodhana', 'cousin', 'yudhishthira'),
    ('nakula', 'cousin', 'duryodhana'), ('duryodhana', 'cousin', 'nakula'),
    ('sahadeva', 'cousin', 'duryodhana'), ('duryodhana', 'cousin', 'sahadeva'),
    ('karna', 'cousin', 'krishna'), ('krishna', 'cousin', 'karna'),  # Added: Karna and Krishna as cousins through maternal lineage

    # In-Law Relations
    ('krishna', 'brother-in-law', 'arjuna'), ('arjuna', 'brother-in-law', 'krishna'),
    ('krishna', 'brother-in-law', 'bhima'), ('bhima', 'brother-in-law', 'krishna'),
    ('krishna', 'brother-in-law', 'yudhishthira'), ('yudhishthira', 'brother-in-law', 'krishna'),
    ('krishna', 'brother-in-law', 'nakula'), ('nakula', 'brother-in-law', 'krishna'),
    ('krishna', 'brother-in-law', 'sahadeva'), ('sahadeva', 'brother-in-law', 'krishna'),
    ('pandu', 'father-in-law', 'draupadi'), ('draupadi', 'daughter-in-law', 'pandu'),
    ('kunti', 'mother-in-law', 'draupadi'), ('draupadi', 'daughter-in-law', 'kunti'),
    ('drupada', 'father-in-law', 'yudhishthira'), ('yudhishthira', 'son-in-law', 'drupada'),
    ('drupada', 'father-in-law', 'bhima'), ('bhima', 'son-in-law', 'drupada'),
    ('drupada', 'father-in-law', 'arjuna'), ('arjuna', 'son-in-law', 'drupada'),
    ('drupada', 'father-in-law', 'nakula'), ('nakula', 'son-in-law', 'drupada'),
    ('drupada', 'father-in-law', 'sahadeva'), ('sahadeva', 'son-in-law', 'drupada'),
    ('vasudeva', 'father-in-law', 'arjuna'), ('arjuna', 'son-in-law', 'vasudeva'),
    ('devaki', 'mother-in-law', 'arjuna'), ('arjuna', 'son-in-law', 'devaki'),
    ('virata', 'father-in-law', 'abhimanyu'), ('abhimanyu', 'son-in-law', 'virata'),
    ('kunti', 'mother-in-law', 'subhadra'), ('subhadra', 'daughter-in-law', 'kunti'),
    ('gandhari', 'mother-in-law', 'bhanumati'), ('bhanumati', 'daughter-in-law', 'gandhari'),
    ('dhritarashtra', 'father-in-law', 'bhanumati'), ('bhanumati', 'daughter-in-law', 'dhritarashtra'),

    # Uncle/Aunt and Nephew/Niece Relations
    ('shakuni', 'uncle', 'duryodhana'), ('duryodhana', 'nephew', 'shakuni'),
    ('shakuni', 'uncle', 'dushasana'), ('dushasana', 'nephew', 'shakuni'),
    ('shakuni', 'uncle', 'vikarna'), ('vikarna', 'nephew', 'shakuni'),
    ('shakuni', 'uncle', 'dushala'), ('dushala', 'niece', 'shakuni'),
    ('vikarna', 'uncle', 'laxman'), ('laxman', 'nephew', 'vikarna'),
    ('kunti', 'aunt', 'abhimanyu'), ('abhimanyu', 'nephew', 'kunti'),
    ('vidura', 'uncle', 'arjuna'), ('arjuna', 'nephew', 'vidura'),
    ('vidura', 'uncle', 'bhima'), ('bhima', 'nephew', 'vidura'),
    ('vidura', 'uncle', 'yudhishthira'), ('yudhishthira', 'nephew', 'vidura'),
    ('vidura', 'uncle', 'nakula'), ('nakula', 'nephew', 'vidura'),
    ('vidura', 'uncle', 'sahadeva'), ('sahadeva', 'nephew', 'vidura'),
    ('vidura', 'uncle', 'duryodhana'), ('duryodhana', 'nephew', 'vidura'),
    ('balarama', 'uncle', 'abhimanyu'), ('abhimanyu', 'nephew', 'balarama'),
    ('krishna', 'uncle', 'parikshit'), ('parikshit', 'nephew', 'krishna'),
    ('krishna', 'uncle', 'abhimanyu'), ('abhimanyu', 'nephew', 'krishna'),
    ('subhadra', 'aunt', 'parikshit'), ('parikshit', 'niece', 'subhadra'),

    # Teacher-Disciple Relations
    ('drona', 'teacher', 'arjuna'), ('arjuna', 'disciple', 'drona'),
    ('drona', 'teacher', 'bhima'), ('bhima', 'disciple', 'drona'),
    ('drona', 'teacher', 'yudhishthira'), ('yudhishthira', 'disciple', 'drona'),
    ('drona', 'teacher', 'nakula'), ('nakula', 'disciple', 'drona'),
    ('drona', 'teacher', 'sahadeva'), ('sahadeva', 'disciple', 'drona'),
    ('drona', 'teacher', 'asvatthama'), ('asvatthama', 'disciple', 'drona'),
    ('drona', 'teacher', 'duryodhana'), ('duryodhana', 'disciple', 'drona'),
    ('drona', 'teacher', 'dushasana'), ('dushasana', 'disciple', 'drona'),
    ('kripa', 'teacher', 'asvatthama'), ('asvatthama', 'disciple', 'kripa'),
    ('kripa', 'teacher', 'arjuna'), ('arjuna', 'disciple', 'kripa'),
    ('kripa', 'teacher', 'duryodhana'), ('duryodhana', 'disciple', 'kripa'),
    ('kripa', 'teacher', 'bhima'), ('bhima', 'disciple', 'kripa'),
    ('kripa', 'teacher', 'yudhishthira'), ('yudhishthira', 'disciple', 'kripa'),
    ('kripa', 'teacher', 'nakula'), ('nakula', 'disciple', 'kripa'),
    ('kripa', 'teacher', 'sahadeva'), ('sahadeva', 'disciple', 'kripa'),
    ('bhishma', 'teacher', 'arjuna'), ('arjuna', 'disciple', 'bhishma'),
    ('bhishma', 'teacher', 'yudhishthira'), ('yudhishthira', 'disciple', 'bhishma'),
    ('bhishma', 'teacher', 'bhima'), ('bhima', 'disciple', 'bhishma'),
    ('bhishma', 'teacher', 'nakula'), ('nakula', 'disciple', 'bhishma'),
    ('bhishma', 'teacher', 'sahadeva'), ('sahadeva', 'disciple', 'bhishma'),
    ('bhishma', 'teacher', 'duryodhana'), ('duryodhana', 'disciple', 'bhishma'),
    ('krishna', 'teacher', 'arjuna'), ('arjuna', 'disciple', 'krishna'),

    # Friend and Ally Relations
    ('karna', 'friend', 'duryodhana'), ('duryodhana', 'friend', 'karna'),
    ('asvatthama', 'ally', 'duryodhana'), ('duryodhana', 'ally', 'asvatthama'),
    ('krishna', 'friend', 'arjuna'), ('arjuna', 'friend', 'krishna'),
    ('krishna', 'friend', 'yudhishthira'), ('yudhishthira', 'friend', 'krishna'),
    ('bhishma', 'guardian', 'duryodhana'), ('duryodhana', 'ward', 'bhishma'),
    ('bhishma', 'guardian', 'arjuna'), ('arjuna', 'ward', 'bhishma'),
    ('satyaki', 'ally', 'arjuna'), ('arjuna', 'ally', 'satyaki'),
    ('dhrishtadyumna', 'ally', 'yudhishthira'), ('yudhishthira', 'ally', 'dhrishtadyumna'),
    ('krishna', 'ally', 'arjuna'), ('arjuna', 'ally', 'krishna'),
    ('krishna', 'ally', 'bhima'), ('bhima', 'ally', 'krishna'),
    ('shalya', 'ally', 'duryodhana'), ('duryodhana', 'ally', 'shalya'),
    ('jayadratha', 'ally', 'duryodhana'), ('duryodhana', 'ally', 'jayadratha'),
]
single_person_mappings = {
    'son': {
        'Arjuna': ('Kunti', 'mother'), 'Bhima': ('Kunti', 'mother'),
        'Yudhishthira': ('Kunti', 'mother'), 'Nakula': ('Madri', 'mother'),
        'Sahadeva': ('Madri', 'mother'), 'Karna': ('Kunti', 'mother'),
        'Krishna': ('Devaki', 'mother'), 'Balarama': ('Devaki', 'mother'),
        'Duryodhana': ('Dhritarashtra', 'father'), 'Dushasana': ('Dhritarashtra', 'father'),
        'Vikarna': ('Dhritarashtra', 'father'), 'Bhurisrava': ('Somadatta', 'father'),
        'Ashwatthama': ('Drona', 'father'), 'Abhimanyu': ('Arjuna', 'father'),
        'Ghatotkacha': ('Bhima', 'father'), 'Yaudheya': ('Yudhishthira', 'father'),
        'Niramitra': ('Nakula', 'father'), 'Suhotra': ('Sahadeva', 'father'),
        'Vrishasena': ('Karna', 'father'), 'Sushena': ('Karna', 'father'),
        'Iravan': ('Arjuna', 'father'), 'Babhruvahana': ('Arjuna', 'father'),
        'Parikshit': ('Abhimanyu', 'father'), 'Laxman': ('Duryodhana', 'father'),
        'Chitrasena': ('Dushasana', 'father'), 'Chitravarma': ('Vikarna', 'father'),
        'Dhritarashtra': ('Vyasa', 'father'), 'Pandu': ('Vyasa', 'father'),
        'Vidura': ('Vyasa', 'father'), 'Bhishma': ('Shantanu', 'father'),
        'Vyasa': ('Parashara', 'father'), 'Chitrangada': ('Shantanu', 'father'),
        'Vichitravirya': ('Shantanu', 'father'), 'Prativindhya': ('Yudhishthira', 'father'),
        'Sutasoma': ('Bhima', 'father'), 'Shrutakarma': ('Arjuna', 'father'),
        'Shrutasena': ('Nakula', 'father'), 'Shrutakirti': ('Sahadeva', 'father'),
        'Partha': ('Kunti', 'mother'),  # From first dictionary
    },
    'daughter': {
        'Draupadi': ('Drupada', 'father'), 'Subhadra': ('Vasudeva', 'father'),
        'Uttara': ('Virata', 'father'), 'Dushala': ('Dhritarashtra', 'father'),
    },
    'father': {
        'Pandu': ('Arjuna', 'son'), 'Dhritarashtra': ('Duryodhana', 'son'),
        'Drupada': ('Draupadi', 'daughter'), 'Vasudeva': ('Krishna', 'son'),
        'Somadatta': ('Bhurisrava', 'son'), 'Drona': ('Ashwatthama', 'son'),
        'Shantanu': ('Bhishma', 'son'), 'Parashara': ('Vyasa', 'son'),
        'Vyasa': ('Dhritarashtra', 'son'), 'Vichitravirya': ('Dhritarashtra', 'son'),
        'Arjuna': ('Abhimanyu', 'son'), 'Bhima': ('Ghatotkacha', 'son'),
        'Yudhishthira': ('Yaudheya', 'son'), 'Nakula': ('Niramitra', 'son'),
        'Sahadeva': ('Suhotra', 'son'), 'Karna': ('Vrishasena', 'son'),
        'Abhimanyu': ('Parikshit', 'son'), 'Duryodhana': ('Laxman', 'son'),
        'Dushasana': ('Chitrasena', 'son'), 'Vikarna': ('Chitravarma', 'son'),
        'Virata': ('Uttara', 'daughter'),
    },
    'mother': {
        'Kunti': ('Arjuna', 'son'), 'Madri': ('Nakula', 'son'),
        'Devaki': ('Krishna', 'son'), 'Gandhari': ('Duryodhana', 'son'),
        'Subhadra': ('Abhimanyu', 'son'), 'Hidimbi': ('Ghatotkacha', 'son'),
        'Devika': ('Yaudheya', 'son'), 'Karenumati': ('Niramitra', 'son'),
        'Vijaya': ('Suhotra', 'son'), 'Vrushali': ('Vrishasena', 'son'),
        'Ganga': ('Bhishma', 'son'), 'Satyavati': ('Vyasa', 'son'),
        'Ambika': ('Dhritarashtra', 'son'), 'Ambalika': ('Pandu', 'son'),
        'Draupadi': ('Prativindhya', 'son'), 'Ulupi': ('Iravan', 'son'),
        'Chitrangada': ('Babhruvahana', 'son'), 'Uttara': ('Parikshit', 'son'),
        'Bhanumati': ('Laxman', 'son'), 'Pritha': ('Arjuna', 'son'),  # From first dictionary
        'Pritha': ('Yudhishthira', 'son'), 'Pritha': ('Karna', 'son'),
    },
    'brother': {
        'Arjuna': ('Bhima', 'brother'), 'Krishna': ('Subhadra', 'sister'),
        'Duryodhana': ('Dushasana', 'brother'), 'Dhrishtadyumna': ('Draupadi', 'sister'),
        'Dhritarashtra': ('Pandu', 'brother'), 'Balarama': ('Krishna', 'brother'),
        'Chitrangada': ('Vichitravirya', 'brother'), 'Arjuna': ('Yudhishthira', 'brother'),
        'Arjuna': ('Nakula', 'brother'), 'Arjuna': ('Sahadeva', 'brother'),
        'Arjuna': ('Karna', 'brother'), 'Bhima': ('Yudhishthira', 'brother'),
        'Bhima': ('Nakula', 'brother'), 'Bhima': ('Sahadeva', 'brother'),
        'Bhima': ('Karna', 'brother'), 'Yudhishthira': ('Nakula', 'brother'),
        'Yudhishthira': ('Sahadeva', 'brother'), 'Yudhishthira': ('Karna', 'brother'),
        'Nakula': ('Sahadeva', 'brother'), 'Nakula': ('Karna', 'brother'),
        'Sahadeva': ('Karna', 'brother'), 'Duryodhana': ('Vikarna', 'brother'),
        'Dushasana': ('Vikarna', 'brother'), 'Pandu': ('Vidura', 'brother'),
        'Partha': ('Bhima', 'brother'),  # From first dictionary
        'Partha': ('Yudhishthira', 'brother'), 'Partha': ('Nakula', 'brother'),
        'Partha': ('Sahadeva', 'brother'), 'Partha': ('Karna', 'brother'),
    },
    'sister': {
        'Subhadra': ('Krishna', 'brother'), 'Draupadi': ('Dhrishtadyumna', 'brother'),
        'Dushala': ('Duryodhana', 'brother'), 'Dushala': ('Dushasana', 'brother'),
        'Dushala': ('Vikarna', 'brother'),
    },
    'husband': {
        'Arjuna': ('Subhadra', 'wife'), 'Yudhishthira': ('Draupadi', 'wife'),
        'Bhima': ('Hidimbi', 'wife'), 'Duryodhana': ('Bhanumati', 'wife'),
        'Karna': ('Vrushali', 'wife'), 'Abhimanyu': ('Uttara', 'wife'),
        'Shantanu': ('Ganga', 'wife'), 'Pandu': ('Kunti', 'wife'),
        'Vasudeva': ('Devaki', 'wife'), 'Jayadratha': ('Dushala', 'wife'),
        'Arjuna': ('Draupadi', 'wife'), 'Arjuna': ('Ulupi', 'wife'),
        'Arjuna': ('Chitrangada', 'wife'), 'Bhima': ('Draupadi', 'wife'),
        'Yudhishthira': ('Devika', 'wife'), 'Nakula': ('Draupadi', 'wife'),
        'Nakula': ('Karenumati', 'wife'), 'Sahadeva': ('Draupadi', 'wife'),
        'Sahadeva': ('Vijaya', 'wife'),
    },
    'wife': {
        'Subhadra': ('Arjuna', 'husband'), 'Draupadi': ('Yudhishthira', 'husband'),
        'Hidimbi': ('Bhima', 'husband'), 'Bhanumati': ('Duryodhana', 'husband'),
        'Vrushali': ('Karna', 'husband'), 'Uttara': ('Abhimanyu', 'husband'),
        'Ganga': ('Shantanu', 'husband'), 'Kunti': ('Pandu', 'husband'),
        'Devaki': ('Vasudeva', 'husband'), 'Dushala': ('Jayadratha', 'husband'),
        'Draupadi': ('Arjuna', 'husband'), 'Draupadi': ('Bhima', 'husband'),
        'Draupadi': ('Nakula', 'husband'), 'Draupadi': ('Sahadeva', 'husband'),
        'Ulupi': ('Arjuna', 'husband'), 'Chitrangada': ('Arjuna', 'husband'),
        'Devika': ('Yudhishthira', 'husband'), 'Karenumati': ('Nakula', 'husband'),
        'Vijaya': ('Sahadeva', 'husband'),
    },
    'brother-in-law': {
        'Krishna': ('Arjuna', 'brother-in-law'), 'Dhrishtadyumna': ('Yudhishthira', 'brother-in-law'),
        'Krishna': ('Bhima', 'brother-in-law'), 'Krishna': ('Yudhishthira', 'brother-in-law'),
        'Krishna': ('Nakula', 'brother-in-law'), 'Krishna': ('Sahadeva', 'brother-in-law'),
        'Arjuna': ('Krishna', 'brother-in-law'), 'Bhima': ('Krishna', 'brother-in-law'),
        'Yudhishthira': ('Krishna', 'brother-in-law'), 'Nakula': ('Krishna', 'brother-in-law'),
        'Sahadeva': ('Krishna', 'brother-in-law'), 'Yudhishthira': ('Dhrishtadyumna', 'brother-in-law'),
    },
    'sister-in-law': {
        'Subhadra': ('Yudhishthira', 'brother-in-law'), 'Draupadi': ('Subhadra', 'sister-in-law'),
        'Subhadra': ('Draupadi', 'sister-in-law'), 'Bhanumati': ('Draupadi', 'sister-in-law'),
    },
    'cousin': {
        'Krishna': ('Arjuna', 'cousin'), 'Balarama': ('Arjuna', 'cousin'),
        'Arjuna': ('Krishna', 'cousin'), 'Arjuna': ('Balarama', 'cousin'),
        'Arjuna': ('Duryodhana', 'cousin'), 'Duryodhana': ('Arjuna', 'cousin'),
        'Krishna': ('Bhima', 'cousin'), 'Krishna': ('Yudhishthira', 'cousin'),
        'Krishna': ('Nakula', 'cousin'), 'Krishna': ('Sahadeva', 'cousin'),
        'Bhima': ('Krishna', 'cousin'), 'Yudhishthira': ('Krishna', 'cousin'),
        'Nakula': ('Krishna', 'cousin'), 'Sahadeva': ('Krishna', 'cousin'),
        'Partha': ('Krishna', 'cousin'),  # From first dictionary
        'Partha': ('Balarama', 'cousin'),
    },
    'grandfather': {
        'Arjuna': ('Parikshit', 'grandson'), 'Bhima': ('Sutasoma', 'grandson'),
        'Yudhishthira': ('Prativindhya', 'grandson'), 'Vyasa': ('Arjuna', 'grandson'),
        'Arjuna': ('Bhishma', 'grandfather'), 'Duryodhana': ('Bhishma', 'grandfather'),
        'Shantanu': ('Dhritarashtra', 'grandson'),
    },
    'grandmother': {
        'Subhadra': ('Parikshit', 'grandson'), 'Draupadi': ('Prativindhya', 'grandson'),
        'Kunti': ('Abhimanyu', 'grandson'), 'Gandhari': ('Laxman', 'grandson'),
        'Bhanumati': ('Laxman', 'grandson'),
        'Satyavati': ('Dhritarashtra', 'great-great-grandson'),
        'Devaki': ('Parikshit', 'great-grandson'),
    },
    'grandson': {
        'Parikshit': ('Arjuna', 'grandfather'), 'Sutasoma': ('Bhima', 'grandfather'),
        'Prativindhya': ('Yudhishthira', 'grandfather'), 'Arjuna': ('Vyasa', 'grandfather'),
        'Abhimanyu': ('Bhishma', 'grandfather'), 'Laxman': ('Duryodhana', 'grandfather'),
        'Dhritarashtra': ('Shantanu', 'grandfather'),
    },
    'great-grandfather': {
        'Bhishma': ('Arjuna', 'great-grandson'),
    },
    'great-grandson': {
        'Arjuna': ('Bhishma', 'great-grandfather'), 'Parikshit': ('Vasudeva', 'great-grandfather'),
    },
    'great-great-grandfather': {
        'Shantanu': ('Dhritarashtra', 'great-great-grandson'),
    },
    'great-great-grandson': {
        'Dhritarashtra': ('Shantanu', 'great-great-grandfather'),
    },
    'uncle': {
        'Shakuni': ('Duryodhana', 'nephew'), 'Vidura': ('Arjuna', 'nephew'),
        'Krishna': ('Abhimanyu', 'nephew'), 'Balarama': ('Abhimanyu', 'nephew'),
    },
    'aunt': {
        'Kunti': ('Abhimanyu', 'nephew'), 'Subhadra': ('Parikshit', 'niece'),
        'Gandhari': ('Arjuna', 'nephew'),
    },
    'nephew': {
        'Duryodhana': ('Shakuni', 'uncle'), 'Arjuna': ('Vidura', 'uncle'),
        'Abhimanyu': ('Krishna', 'uncle'), 'Laxman': ('Vikarna', 'uncle'),
    },
    'niece': {
        'Parikshit': ('Subhadra', 'aunt'), 'Dushala': ('Shakuni', 'uncle'),
        'Uttara': ('Krishna', 'brother-in-law'),
    },
    'father-in-law': {
        'Drupada': ('Yudhishthira', 'son-in-law'), 'Vasudeva': ('Arjuna', 'son-in-law'),
        'Virata': ('Abhimanyu', 'son-in-law'), 'Pandu': ('Draupadi', 'daughter-in-law'),
        'Dhritarashtra': ('Bhanumati', 'daughter-in-law'),
    },
    'mother-in-law': {
        'Kunti': ('Draupadi', 'daughter-in-law'), 'Devaki': ('Arjuna', 'son-in-law'),
        'Gandhari': ('Bhanumati', 'daughter-in-law'),
    },
    'son-in-law': {
        'Yudhishthira': ('Drupada', 'father-in-law'), 'Arjuna': ('Vasudeva', 'father-in-law'),
        'Abhimanyu': ('Virata', 'father-in-law'),
    },
    'daughter-in-law': {
        'Draupadi': ('Kunti', 'mother-in-law'), 'Subhadra': ('Kunti', 'mother-in-law'),
        'Bhanumati': ('Gandhari', 'mother-in-law'),
    },
    'friend': {
        'Arjuna': ('Krishna', 'friend'), 'Karna': ('Duryodhana', 'friend'),
        'Yudhishthira': ('Krishna', 'friend'),
        'Partha': ('Krishna', 'friend'),  # From first dictionary
    },
    'teacher': {
        'Drona': ('Arjuna', 'disciple'), 'Kripa': ('Duryodhana', 'disciple'),
        'Bhishma': ('Arjuna', 'disciple'), 'Krishna': ('Arjuna', 'disciple'),
        'Drona': ('Ashwatthama', 'disciple'),
    },
    'disciple': {
        'Arjuna': ('Drona', 'teacher'), 'Duryodhana': ('Kripa', 'teacher'),
        'Arjuna': ('Bhishma', 'teacher'), 'Arjuna': ('Krishna', 'teacher'),
        'Ashwatthama': ('Drona', 'teacher'),
        'Partha': ('Drona', 'teacher'),  # From first dictionary
        'Partha': ('Bhishma', 'teacher'), 'Partha': ('Krishna', 'teacher'),
    },
    'ally': {
        'Satyaki': ('Arjuna', 'ally'), 'Dhrishtadyumna': ('Yudhishthira', 'ally'),
        'Ashwatthama': ('Duryodhana', 'ally'), 'Krishna': ('Arjuna', 'ally'),
        'Shalya': ('Duryodhana', 'ally'), 'Jayadratha': ('Duryodhana', 'ally'),
    },
    'guardian': {
        'Bhishma': ('Duryodhana', 'ward'), 'Bhishma': ('Arjuna', 'ward'),
    },
    'ward': {
        'Duryodhana': ('Bhishma', 'guardian'), 'Arjuna': ('Bhishma', 'guardian'),
        'Partha': ('Bhishma', 'guardian'),  # From first dictionary
    },
}
# Expanded invalid_pairs to include incorrect Krishna-Bhima brother relation
invalid_pairs = {
    # Incorrect parent-child relations
    ('Brahma', 'son', 'Kunti'), ('Brahma', 'daughter', 'Kunti'),
    ('Bhishma', 'son', 'Pritha'), ('Bhishma', 'daughter', 'Pritha'),
    ('Yodhamanyu', 'son', 'Subhadra'), ('Yodhamanyu', 'son', 'Draupadi'),
    ('Kunti', 'son', 'Unknown'), ('Kunti', 'son', 'Brahma'),
    ('Draupadi', 'son', 'Unknown'), ('Subhadra', 'son', 'Unknown'),
    ('Dhritarashtra', 'son', 'Pandu'), ('Madhava', 'son', 'Pandu'),
    ('Drupada', 'son', 'Subhadra'), ('Drupada', 'daughter', 'Subhadra'),
    ('Bhishma', 'son', 'Unknown'), ('Brahma', 'son', 'Unknown'),
    ('Bhishma', 'son', 'Somadatta'), ('Pandu', 'son', 'Drupada'),
    ('Madhava', 'son', 'Unknown'), ('Pritha', 'son', 'Unknown'),
    ('Arjuna', 'son', 'Pandu'), ('Yodhamanyu', 'son', 'Unknown'),
    ('Dhritarashtra', 'son', 'Unknown'), ('Pandu', 'son', 'Unknown'),
    ('Arjuna', 'father', 'Unknown'), ('Drupada', 'son', 'Draupadi'),
    ('Draupadi', 'mother', 'Drupada'), ('Krishna', 'son', 'Pandu'),
    ('Drupada', 'son', 'Unknown'), ('Kunti', 'father', 'Unknown'),
    ('Vasuki', 'father', 'Unknown'), ('Vayu', 'grandfather', 'Unknown'),
    ('Krishna', 'father', 'Draupadi'), ('Krishna', 'grandfather', 'Draupadi'),
    ('Krishna', 'mother', 'Draupadi'), ('Krishna', 'grandmother', 'Draupadi'),
    ('Arjuna', 'father', 'Kunti'), ('Kunti', 'daughter', 'Arjuna'),
    ('Draupadi', 'father', 'Yudhishthira'), ('Yudhishthira', 'mother', 'Draupadi'),
    ('Bhishma', 'father', 'Arjuna'), ('Duryodhana', 'brother', 'Arjuna'),
    ('Arjuna', 'brother', 'Duryodhana'), ('Krishna', 'son', 'Arjuna'),
    ('Arjuna', 'father', 'Krishna'), ('Subhadra', 'mother', 'Krishna'),
    ('Krishna', 'father', 'Subhadra'), ('Draupadi', 'sister', 'Arjuna'),
    ('Arjuna', 'brother', 'Draupadi'), ('Karna', 'father', 'Kunti'),
    ('Kunti', 'daughter', 'Karna'), ('Abhimanyu', 'father', 'Arjuna'),
    ('Arjuna', 'son', 'Abhimanyu'), ('Parikshit', 'father', 'Abhimanyu'),
    ('Abhimanyu', 'grandfather', 'Parikshit'), ('Gandhari', 'daughter', 'Duryodhana'),
    ('Duryodhana', 'father', 'Gandhari'), ('Shakuni', 'father', 'Duryodhana'),
    ('Duryodhana', 'nephew', 'Kunti'), ('Drona', 'father', 'Arjuna'),
    ('Arjuna', 'son', 'Drona'), ('Ashwatthama', 'brother', 'Arjuna'),
    ('Draupadi', 'mother', 'Krishna'), ('Krishna', 'husband', 'Draupadi'),
    ('Subhadra', 'wife', 'Yudhishthira'), ('Uttara', 'mother', 'Arjuna'),
    ('Arjuna', 'son', 'Uttara'), ('Virata', 'father', 'Arjuna'),
    ('Arjuna', 'son', 'Virata'), ('Dushala', 'brother', 'Arjuna'),
    ('Arjuna', 'sister', 'Dushala'), ('Vyasa', 'son', 'Dhritarashtra'),
    ('Dhritarashtra', 'father', 'Vyasa'), ('Shantanu', 'son', 'Bhishma'),
    ('Bhishma', 'father', 'Shantanu'), ('Ganga', 'daughter', 'Bhishma'),
    ('Bhishma', 'father', 'Ganga'), ('Satyavati', 'daughter', 'Vyasa'),
    ('Vyasa', 'father', 'Satyavati'), ('Dhrishtadyumna', 'husband', 'Draupadi'),
    ('Draupadi', 'wife', 'Dhrishtadyumna'), ('Kripa', 'father', 'Arjuna'),
    ('Arjuna', 'son', 'Kripa'), ('Bhanumati', 'sister', 'Duryodhana'),
    ('Duryodhana', 'brother', 'Bhanumati'), ('Vrushali', 'sister', 'Karna'),
    ('Karna', 'brother', 'Vrushali'), ('Hidimbi', 'sister', 'Bhima'),
    ('Bhima', 'brother', 'Hidimbi'), ('Ulupi', 'sister', 'Arjuna'),
    ('Arjuna', 'brother', 'Ulupi'), ('Chitrangada', 'sister', 'Arjuna'),
    ('Arjuna', 'brother', 'Chitrangada'), ('Devika', 'sister', 'Yudhishthira'),
    ('Yudhishthira', 'brother', 'Devika'), ('Karenumati', 'sister', 'Nakula'),
    ('Nakula', 'brother', 'Karenumati'), ('Vijaya', 'sister', 'Sahadeva'),
    ('Sahadeva', 'brother', 'Vijaya'), ('Drona', 'brother', 'Arjuna'),
    ('Arjuna', 'brother', 'Drona'), ('Vasudeva', 'son', 'Krishna'),
    ('Krishna', 'father', 'Vasudeva'), ('Devaki', 'daughter', 'Krishna'),
    ('Krishna', 'father', 'Devaki'), ('Nakula', 'father', 'Arjuna'),
    ('Arjuna', 'son', 'Nakula'), ('Sahadeva', 'father', 'Bhima'),
    ('Bhima', 'son', 'Sahadeva'), ('Yudhishthira', 'father', 'Arjuna'),
    ('Arjuna', 'son', 'Yudhishthira'), ('Karna', 'brother', 'Duryodhana'),
    ('Duryodhana', 'brother', 'Karna'), ('Subhadra', 'sister', 'Draupadi'),
    ('Draupadi', 'sister', 'Subhadra'), ('Bhishma', 'brother', 'Pandu'),
    ('Pandu', 'brother', 'Bhishma'), ('Ghatotkacha', 'father', 'Bhima'),
    ('Bhima', 'son', 'Ghatotkacha'), ('Yaudheya', 'father', 'Yudhishthira'),
    ('Yudhishthira', 'son', 'Yaudheya'), ('Niramitra', 'father', 'Nakula'),
    ('Nakula', 'son', 'Niramitra'), ('Suhotra', 'father', 'Sahadeva'),
    ('Sahadeva', 'son', 'Suhotra'), ('Iravan', 'father', 'Arjuna'),
    ('Arjuna', 'son', 'Iravan'), ('Babhruvahana', 'father', 'Arjuna'),
    ('Arjuna', 'son', 'Babhruvahana'), ('Vrishasena', 'father', 'Karna'),
    ('Karna', 'son', 'Vrishasena'), ('Sushena', 'father', 'Karna'),
    ('Karna', 'son', 'Sushena'), ('Laxman', 'father', 'Duryodhana'),
    ('Duryodhana', 'son', 'Laxman'), ('Chitrasena', 'father', 'Dushasana'),
    ('Dushasana', 'son', 'Chitrasena'), ('Chitravarma', 'father', 'Vikarna'),
    ('Vikarna', 'son', 'Chitravarma'), ('Arjuna', 'grandfather', 'Abhimanyu'),
    ('Abhimanyu', 'grandson', 'Arjuna'), ('Bhima', 'grandfather', 'Ghatotkacha'),
    ('Ghatotkacha', 'grandson', 'Bhima'), ('Drona', 'father', 'Duryodhana'),
    ('Duryodhana', 'son', 'Drona'), ('Kripa', 'father', 'Duryodhana'),
    ('Duryodhana', 'son', 'Kripa'), ('Shalya', 'father', 'Arjuna'),
    ('Arjuna', 'son', 'Shalya'), ('Satyaki', 'brother', 'Arjuna'),
    ('Arjuna', 'brother', 'Satyaki'), ('Jayadratha', 'brother', 'Duryodhana'),
    ('Duryodhana', 'brother', 'Jayadratha'), ('Krishna', 'father', 'Abhimanyu'),
    ('Abhimanyu', 'son', 'Krishna'), ('Subhadra', 'mother', 'Parikshit'),
    ('Parikshit', 'son', 'Subhadra'), ('Uttara', 'wife', 'Arjuna'),
    ('Arjuna', 'husband', 'Uttara'), ('Bhanumati', 'mother', 'Duryodhana'),
    ('Duryodhana', 'father', 'Bhanumati'), ('Vrushali', 'mother', 'Karna'),
    ('Karna', 'father', 'Vrushali'), ('Hidimbi', 'mother', 'Bhima'),
    ('Bhima', 'father', 'Hidimbi'), ('Ulupi', 'mother', 'Arjuna'),
    ('Arjuna', 'father', 'Ulupi'), ('Chitrangada', 'mother', 'Arjuna'),
    ('Arjuna', 'father', 'Chitrangada'), ('Devika', 'mother', 'Yudhishthira'),
    ('Yudhishthira', 'father', 'Devika'), ('Karenumati', 'mother', 'Nakula'),
    ('Nakula', 'father', 'Karenumati'), ('Vijaya', 'mother', 'Sahadeva'),
    ('Sahadeva', 'father', 'Vijaya'), ('Draupadi', 'daughter', 'Yudhishthira'),
    ('Yudhishthira', 'father', 'Draupadi'), ('Subhadra', 'daughter', 'Arjuna'),
    ('Arjuna', 'father', 'Subhadra'), ('Bhishma', 'grandfather', 'Kunti'),
    ('Kunti', 'grandson', 'Bhishma'), ('Shantanu', 'father', 'Pandu'),
    ('Pandu', 'son', 'Shantanu'), ('Vyasa', 'grandfather', 'Arjuna'),
    ('Arjuna', 'grandson', 'Vyasa'), ('Drona', 'teacher', 'Krishna'),
    ('Krishna', 'disciple', 'Drona'), ('Kripa', 'teacher', 'Krishna'),
    ('Krishna', 'disciple', 'Kripa'), ('Vidura', 'father', 'Arjuna'),
    ('Arjuna', 'son', 'Vidura'), ('Vidura', 'brother', 'Arjuna'),
    ('Arjuna', 'brother', 'Vidura'), ('Sahadeva', 'husband', 'Arjuna'),
    ('Arjuna', 'wife', 'Sahadeva'), ('Yudhishthira', 'brother', 'Abhimanyu'),
    ('Abhimanyu', 'brother', 'Yudhishthira'), ('Nakula', 'brother', 'Ghatotkacha'),
    ('Ghatotkacha', 'brother', 'Nakula'), ('Dushasana', 'brother', 'Karna'),
    ('Karna', 'brother', 'Dushasana'), ('Subhadra', 'sister', 'Uttara'),
    ('Uttara', 'sister', 'Subhadra'), ('Bhima', 'father', 'Abhimanyu'),
    ('Abhimanyu', 'son', 'Bhima'), ('Yudhishthira', 'father', 'Iravan'),
    ('Iravan', 'son', 'Yudhishthira'), ('Duryodhana', 'father', 'Vrishasena'),
    ('Vrishasena', 'son', 'Duryodhana'), ('Gandhari', 'mother', 'Draupadi'),
    ('Draupadi', 'daughter', 'Gandhari'), ('Subhadra', 'mother', 'Laxman'),
    ('Laxman', 'son', 'Subhadra'), ('Yudhishthira', 'husband', 'Subhadra'),
    ('Subhadra', 'wife', 'Yudhishthira'), ('Nakula', 'husband', 'Uttara'),
    ('Uttara', 'wife', 'Nakula'), ('Duryodhana', 'husband', 'Vrushali'),
    ('Vrushali', 'wife', 'Duryodhana'), ('Karna', 'husband', 'Bhanumati'),
    ('Bhanumati', 'wife', 'Karna'), ('Arjuna', 'grandfather', 'Sutasoma'),
    ('Sutasoma', 'grandson', 'Arjuna'), ('Bhima', 'grandfather', 'Shrutakarma'),
    ('Shrutakarma', 'grandson', 'Bhima'), ('Duryodhana', 'grandfather', 'Chitrasena'),
    ('Chitrasena', 'grandson', 'Duryodhana'), ('Kunti', 'grandmother', 'Yaudheya'),
    ('Yaudheya', 'grandson', 'Kunti'), ('Yudhishthira', 'teacher', 'Drona'),
    ('Drona', 'disciple', 'Yudhishthira'), ('Duryodhana', 'teacher', 'Kripa'),
    ('Kripa', 'disciple', 'Duryodhana'), ('Abhimanyu', 'teacher', 'Krishna'),
    ('Krishna', 'disciple', 'Abhimanyu'), ('Arjuna', 'cousin', 'Ghatotkacha'),
    ('Ghatotkacha', 'cousin', 'Arjuna'), ('Duryodhana', 'cousin', 'Parikshit'),
    ('Parikshit', 'cousin', 'Duryodhana'), ('Subhadra', 'mother-in-law', 'Draupadi'),
    ('Draupadi', 'daughter-in-law', 'Subhadra'), ('Gandhari', 'mother-in-law', 'Subhadra'),
    ('Subhadra', 'daughter-in-law', 'Gandhari'), ('Krishna', 'brother-in-law', 'Duryodhana'),
    ('Duryodhana', 'brother-in-law', 'Krishna'), ('Indra', 'father', 'Arjuna'),
    ('Arjuna', 'son', 'Indra'), ('Yama', 'father', 'Yudhishthira'),
    ('Yudhishthira', 'son', 'Yama'), ('Unknown', 'father', 'Krishna'),
    ('Krishna', 'son', 'Unknown'), ('Parikshit', 'father', 'Yudhishthira'),
    ('Yudhishthira', 'son', 'Parikshit'), ('Abhimanyu', 'father', 'Arjuna'),
    ('Arjuna', 'son', 'Abhimanyu'), ('Vyasa', 'father', 'Shantanu'),
    ('Shantanu', 'son', 'Vyasa'), ('Arjuna', 'son', 'Vyasa'),
    ('Yudhishthira', 'son', 'Shantanu'), ('Duryodhana', 'son', 'Bhishma'),
    ('Parikshit', 'son', 'Yudhishthira'), ('Arjuna', 'teacher', 'Drona'),
    ('Drona', 'disciple', 'Arjuna'), ('Krishna', 'disciple', 'Bhishma'),
    ('Bhishma', 'teacher', 'Krishna'), ('Krishna', 'brother', 'Bhima'),
    ('Bhima', 'brother', 'Krishna'), ('Balarama', 'brother', 'Arjuna'),
    ('Arjuna', 'brother', 'Balarama'),

    # Incorrect spouse relations
    ('Bhishma', 'husband', 'Gandhari'), ('Kunti', 'wife', 'Pandu'),
    ('Pandu', 'husband', 'Kunti'), ('Gandhari', 'wife', 'Dhritarashtra'),
    ('Dhritarashtra', 'husband', 'Gandhari'), ('Devaki', 'wife', 'Vasudeva'),
    ('Vasudeva', 'husband', 'Devaki'), ('Dushala', 'wife', 'Jayadratha'),
    ('Jayadratha', 'husband', 'Dushala'),

    # Incorrect cousin relations
    ('Arjuna', 'cousin', 'Duryodhana'), ('Duryodhana', 'cousin', 'Arjuna'),
    ('Karna', 'cousin', 'Arjuna'), ('Arjuna', 'cousin', 'Karna'),

    # Incorrect teacher-disciple relations
    ('Krishna', 'teacher', 'Arjuna'), ('Arjuna', 'disciple', 'Krishna'),

    # Incorrect other relations
    ('Arjuna', 'maternal uncle', 'Unknown'), ('Arjuna', 'grandson', 'Unknown'),
    ('Bhishma', 'grandsire', 'Unknown'), ('Drupada', 'lord', 'Subhadra'),
    ('Sanjaya', 'teacher', 'Drona'), ('Sanjaya', 'lord', 'Arjuna'),
    ('Sanjaya', 'foes', 'Krishna'), ('Sanjaya', 'lord', 'Krishna'),
    ('Krishna', 'lord', 'Arjuna'), ('Duryodhana', 'son', 'Kripa'),
    ('Duryodhana', 'husband', 'Kripa'), ('Duryodhana', 'family', 'Kripa'),
    ('Arjuna', 'father', 'Balarama'), ('Arjuna', 'mother', 'Balarama'),
    ('Arjuna', 'family', 'Balarama'), ('Uttara', 'mother', 'Abhimanyu'),
    ('Satyavati', 'son', 'Vyasa'), ('Chitrangada', 'son', 'Babhruvahana'),
    ('Bhanumati', 'father', 'Laxman'),
    ('dhritarashtra', 'son', 'pandu'), ('pandu', 'son', 'drupada'), ('yodhamanyu', 'son', 'subhadra'),
    ('yodhamanyu', 'son', 'draupadi'), ('bhishma', 'son', 'somadatta'), ('krishna', 'son', 'pandu'),
    ('drupada', 'son', 'draupadi'), ('drupada', 'lord', 'subhadra'), ('drupada', 'son', 'subhadra'),
    ('brahma', 'son', 'kunti'), ('dhritarashtra', 'family', 'krishna'), ('sanjaya', 'lord', 'arjuna'),
    ('sanjaya', 'foes', 'krishna'), ('sanjaya', 'lord', 'krishna'), ('krishna', 'lord', 'arjuna'),
    ('yodhamanyu', 'son', 'vasudeva (father)'), ('uttara', 'sister', 'parikshit'),
    ('ulupi', 'mother', 'vidura'), ('arjuna', 'son', 'srinjaya'), ('bhanumati', 'mother', 'shrutasena'),
    ('shesha', 'son', 'kekaya'), ('satyavati', 'mother', 'brihadkshatra'), ('kripa', 'son', 'vasudeva (father)'),
    ('arjuna', 'great-grandson', 'bhishma'), ('dhritarashtra', 'great-great-grandson', 'satyavati'),
    ('gandhari', 'sister', 'prativindhya'), ('asvatthama', 'son', 'gandhari'),
    ('ambalika', 'mother', 'charudeshna'), ('chitrangada', 'mother', 'brihadkshatra'),
    ('devika', 'mother', 'vidura'), ('dushala', 'mother', 'ugrasena'),
    ('vikarna', 'brother', 'sini'), ('vikarna', 'great-grandson', 'kapila'), ('vikarna', 'friend', 'ambalika'),
    ('vikarna', 'grandson', 'draupadi'), ('vikarna', 'great-grandson', 'subhadra'), ('vikarna', 'great-grandson', 'ghatotkacha'),
    ('vikarna', 'son-in-law', 'rukmini'), ('vikarna', 'great-grandson', 'shrutasena'), ('vikarna', 'uncle', 'ugrasena'),
    ('vikarna', 'nephew', 'satyajit'), ('vikarna', 'ward', 'devaki'), ('vikarna', 'father-in-law', 'shantanu'),
    ('vikarna', 'ward', 'kashyapa'), ('vikarna', 'ward', 'drona'), ('vikarna', 'great-grandfather', 'chitrasena'),
    ('vikarna', 'friend', 'amba'), ('vikarna', 'son-in-law', 'abhimanyu'), ('vikarna', 'disciple', 'ganga'),
    ('vikarna', 'disciple', 'sikhandi'), ('vikarna', 'brother-in-law', 'kekaya'), ('vikarna', 'brother', 'vasudeva (father)'),
    ('vikarna', 'guardian', 'kripa'), ('vikarna', 'cousin', 'sutasoma'), ('vikarna', 'uncle', 'krishna'),
    ('vikarna', 'ally', 'parashara'), ('vikarna', 'disciple', 'saurin'), ('vikarna', 'teacher', 'shrutayudha'),
    ('vikarna', 'son-in-law', 'vichitravirya'), ('vikarna', 'great-great-grandson', 'hidimbi'), ('vikarna', 'friend', 'jayadratha'),
    ('vikarna', 'friend', 'srinjaya'), ('vikarna', 'brother-in-law', 'vidura'), ('vikarna', 'disciple', 'nakula'),
    ('vikarna', 'great-grandfather', 'narada'), ('vikarna', 'great-grandson', 'gandhari'), ('vikarna', 'father', 'vrishasena')

}

# Step 7: Optimized Helper Functions for Code 2
def clean_text(text):
    text = unicodedata.normalize('NFKD', str(text)).encode('ascii', 'ignore').decode('ascii')
    text = re.sub(r'[^\x20-\x7E]', ' ', text)
    return text.strip() or "empty"

def tokenize(text):
    tokens = []
    text = clean_text(text)
    doc = nlp(text)
    i = 0
    while i < len(doc):
        matched = False
        for j in range(5, 0, -1):  # Increased window to 5 for multi-token names
            if i + j <= len(doc):
                candidate = ' '.join(t.text for t in doc[i:i+j])
                candidate_lower = candidate.lower()
                if (candidate_lower in person_names_set or
                    candidate_lower in locations_set or
                    candidate_lower in relation_keywords_set):
                    tokens.append(candidate)
                    i += j
                    matched = True
                    break
        if not matched:
            token = doc[i]
            token_lower = token.text.lower()
            if (not token.is_stop or
                token_lower in person_names_set or
                token_lower in locations_set or
                token_lower in relation_keywords_set):
                tokens.append(token.text)
            i += 1
    return [t for t in tokens if t]

def tag_tokens(tokens):
    tagged = []
    i = 0
    while i < len(tokens):
        token = tokens[i]
        token_lower = token.lower()
        if token_lower in person_names_set:
            tagged.append((token, 'B-PER'))
        elif token_lower in locations_set:
            tagged.append((token, 'B-LOC'))
        elif token_lower in relation_keywords_set:
            tagged.append((token, 'B-REL'))
            j = i + 1
            while j < len(tokens):
                combined = ' '.join(tokens[i:j+1]).lower()
                if combined in relation_keywords_set:
                    tagged.append((tokens[j], 'I-REL'))
                    j += 1
                else:
                    break
            i = j - 1
        else:
            tagged.append((token, 'O'))
        i += 1
    return tagged

# Updated extract_relations with contextual rules for rare relations
def extract_relations(tagged_tokens, chapter, verse):
    relations = []
    untagged_keywords = []
    potential_relations = []
    predefined_family_pairs = {(r[0], r[2]) for r in predefined_relations if r[1] == 'family'}

    # Extract persons from tokens
    persons = []
    for i, (token, tag) in enumerate(tagged_tokens):
        if tag == 'B-PER':
            token_lower = token.lower()
            canonical_name = name_to_canonical.get(token_lower, token_lower)
            if canonical_name in person_names_set:
                persons.append((i, canonical_name))

    # Use predefined relations
    for i, p1 in enumerate(persons):
        idx1, person1 = p1
        for j, p2 in enumerate(persons[i+1:]):
            idx2, person2 = p2
            for rel in ['son', 'father', 'mother', 'brother', 'family', 'cousin', 'brother-in-law', 'teacher', 'disciple', 'friend', 'ally', 'wife', 'husband']:
                if (person1, rel, person2) in predefined_relations and (person1, rel, person2) not in invalid_pairs:
                    relations.append((person1, rel, person2))
                    reverse_rel = single_person_mappings.get(rel, {}).get(person2, (None, None))[1]
                    if reverse_rel and (person2, reverse_rel, person1) not in invalid_pairs:
                        relations.append((person2, reverse_rel, person1))

    # Keyword-based extraction with contextual rules
    for i, (token, tag) in enumerate(tagged_tokens):
        if tag == 'B-REL':
            keyword = token.lower()
            # Normalize keywords
            if keyword in ['sons', 'child', 'children', 'offspring', 'son of', 'children of', 'child of', 'born', 'scion', 'issue', 'begotten']:
                keyword = 'son'
            elif keyword in ['daughters', 'daughter of']:
                keyword = 'daughter'
            elif keyword in ['fathers', 'parents', 'ancestors', 'forefather', 'progenitor', 'sire', 'ancestry', 'heritage', 'clan', 'tribe', 'household']:
                keyword = 'father'
            elif keyword in ['brothers', 'siblings', 'brethren']:
                keyword = 'brother'
            elif keyword in ['sisters']:
                keyword = 'sister'
            elif keyword in ['cousins']:
                keyword = 'cousin'
            elif keyword in ['grandfathers', 'grandsons', 'maternal uncles', 'descendants']:
                keyword = keyword.rstrip('s')
            elif keyword in ['kin', 'kinsmen', 'relatives', 'born to', 'born of', 'relative', 'kinsman', 'heir', 'descendant', 'progeny', 'lineage', 'parentage']:
                keyword = 'family'
            elif keyword in ['teacher', 'mentor', 'guru', 'instructor']:
                keyword = 'teacher'
            elif keyword in ['disciple', 'student', 'pupil']:
                keyword = 'disciple'
            elif keyword in ['friend', 'companion', 'ally']:
                keyword = 'friend'

            window = 30
            person_before = None
            person_after = None
            for j in range(max(0, i - window), i):
                if tagged_tokens[j][1] == 'B-PER':
                    person_before = tagged_tokens[j][0].lower()
                    person_before = name_to_canonical.get(person_before, person_before)
                    break
            for j in range(i + 1, min(len(tagged_tokens), i + window + 1)):
                if tagged_tokens[j][1] == 'B-PER':
                    person_after = tagged_tokens[j][0].lower()
                    person_after = name_to_canonical.get(person_after, person_after)
                    break

            # Contextual extraction for teacher-disciple and friend
            if person_before and person_after and person_before != person_after:
                before_gender = gender_map.get(person_before, 'unknown')
                after_gender = gender_map.get(person_after, 'unknown')
                valid_relation = False
                final_keyword = keyword
                relation = (person_before, final_keyword, person_after)

                # Gender and role-based validation
                if keyword == 'son':
                    if before_gender == 'male' and after_gender in ['female', 'male']:
                        valid_relation = True
                    elif after_gender == 'male' and before_gender == 'female':
                        valid_relation = True
                        person_before, person_after = person_after, person_before
                        final_keyword = 'mother'
                        relation = (person_before, final_keyword, person_after)
                elif keyword == 'daughter':
                    if before_gender == 'female' and after_gender == 'male':
                        valid_relation = True
                    elif after_gender == 'female' and before_gender == 'male':
                        valid_relation = True
                        person_before, person_after = person_after, person_before
                        final_keyword = 'father'
                        relation = (person_before, final_keyword, person_after)
                elif keyword == 'father':
                    if before_gender == 'male' and after_gender in ['male', 'female']:
                        valid_relation = True
                elif keyword == 'mother':
                    if before_gender == 'female' and after_gender in ['male', 'female']:
                        valid_relation = True
                elif keyword == 'brother':
                    if before_gender == 'male' and after_gender == 'male':
                        valid_relation = True
                elif keyword == 'sister':
                    if before_gender == 'female' and after_gender == 'female':
                        valid_relation = True
                elif keyword == 'cousin':
                    if before_gender == after_gender:
                        valid_relation = True
                elif keyword == 'brother-in-law':
                    if before_gender == 'male':
                        valid_relation = True
                elif keyword == 'husband':
                    if before_gender == 'male' and after_gender == 'female':
                        valid_relation = True
                elif keyword == 'wife':
                    if before_gender == 'female' and after_gender == 'male':
                        valid_relation = True
                elif keyword == 'family':
                    valid_relation = (person_before, person_after) in predefined_family_pairs
                elif keyword == 'teacher':
                    if before_gender == 'male':  # Teachers are typically male in Mahabharata
                        # Check for contextual clues
                        context = ' '.join(t[0].lower() for t in tagged_tokens[max(0, i-5):min(len(tagged_tokens), i+6)])
                        if 'instruct' in context or 'guide' in context or 'teach' in context:
                            valid_relation = True
                            final_keyword = 'teacher'
                            relation = (person_before, final_keyword, person_after)
                elif keyword == 'disciple':
                    if after_gender in ['male', 'female']:
                        context = ' '.join(t[0].lower() for t in tagged_tokens[max(0, i-5):min(len(tagged_tokens), i+6)])
                        if 'instruct' in context or 'guide' in context or 'teach' in context:
                            valid_relation = True
                            final_keyword = 'disciple'
                            relation = (person_after, final_keyword, person_before)
                elif keyword == 'friend':
                    context = ' '.join(t[0].lower() for t in tagged_tokens[max(0, i-5):min(len(tagged_tokens), i+6)])
                    if 'beloved' in context or 'companion' in context or 'friend' in context:
                        valid_relation = True
                        final_keyword = 'friend'
                        relation = (person_before, final_keyword, person_after)

                if valid_relation and relation not in invalid_pairs:
                    potential_relations.append((relation, 'Valid', before_gender, after_gender))
                    relations.append(relation)
                    # Add reverse relation
                    if final_keyword == 'son' and before_gender == 'male' and after_gender == 'female':
                        reverse = (person_after, 'mother', person_before)
                        if reverse not in invalid_pairs:
                            relations.append(reverse)
                    elif final_keyword == 'daughter' and before_gender == 'female' and after_gender == 'male':
                        reverse = (person_after, 'father', person_before)
                        if reverse not in invalid_pairs:
                            relations.append(reverse)
                    elif final_keyword in ['brother', 'sister', 'cousin', 'brother-in-law', 'friend']:
                        reverse = (person_after, final_keyword, person_before)
                        if reverse not in invalid_pairs:
                            relations.append(reverse)
                    elif final_keyword == 'teacher':
                        reverse = (person_after, 'disciple', person_before)
                        if reverse not in invalid_pairs:
                            relations.append(reverse)
                    elif final_keyword == 'disciple':
                        reverse = (person_before, 'teacher', person_after)
                        if reverse not in invalid_pairs:
                            relations.append(reverse)
                    elif final_keyword == 'husband':
                        reverse = (person_after, 'wife', person_before)
                        if reverse not in invalid_pairs:
                            relations.append(reverse)
                    elif final_keyword == 'wife':
                        reverse = (person_after, 'husband', person_before)
                        if reverse not in invalid_pairs:
                            relations.append(reverse)
                    elif final_keyword == 'family':
                        reverse = (person_after, final_keyword, person_before)
                        if reverse not in invalid_pairs:
                            relations.append(reverse)
                else:
                    potential_relations.append((relation, f'Invalid (Gender: {before_gender}, {after_gender})', before_gender, after_gender))

            if not (person_before or person_after):
                start = max(0, i - 15)
                end = min(len(tagged_tokens), i + 16)
                context = ' '.join(t[0] for t in tagged_tokens[start:end])
                untagged_keywords.append((keyword, chapter, verse, context))

    for relation, status, gender1, gender2 in potential_relations:
        print(f"{status}: {relation[0]} -> {relation[1]} -> {relation[2]} (Gender: {gender1}, {gender2})")

    return relations, untagged_keywords

#Integrate all preprocessing and extraction steps for a single verse.
def process_verse(chapter, verse, text):
    text = clean_text(text)
    if not text:
        return [], [], []
    tokens = tokenize(text)
    tagged = tag_tokens(tokens)
    relations, untagged = extract_relations(tagged, chapter, verse)
    annotated = [(str(chapter), 'O'), ('.', 'O'), (str(verse), 'O')] + tagged
    # Orchestrates preprocessing and extraction, producing all necessary outputs for Cell 3.
    return annotated, relations, untagged

# Step 8: Load and process the CSV for Code 2
try:
    try:
        df = pd.read_csv("Bhagwad_Gita.csv", encoding='utf-8')
    except UnicodeDecodeError:
        df = pd.read_csv("Bhagwad_Gita.csv", encoding='latin1')
    gita_data = df[['Chapter', 'Verse', 'EngMeaning']].dropna()
    gita_data = gita_data.rename(columns={'EngMeaning': 'Text'})
except FileNotFoundError:
    print("Error: 'Bhagwad_Gita.csv' not found.")
    exit(1)
except Exception as e:
    print(f"Error loading CSV: {e}")
    exit(1)

# Log CSV details
print(f"Loaded CSV: {len(gita_data)} rows")
print(f"Columns found: {list(gita_data.columns)}")
print(f"First 5 rows:\n{gita_data.head(5).to_string()}")
print(f"First row: {gita_data.iloc[0][['Chapter', 'Verse', 'Text']].to_dict()}")

print("Setup complete. Run the next cell to process the data and extract relations.")
```

```python
# Cell 2: Process verses, extract relations, clean, balance, augment, and prepare train/dev/test splits

import random
from itertools import combinations
from collections import defaultdict, Counter
from sklearn.model_selection import train_test_split
import os

# Verify dependencies from Cell 1
if 'gita_data' not in globals():
    raise NameError("gita_data must be defined in Cell 1")
if 'person_names' not in globals():
    raise NameError("person_names must be defined in Cell 1")
if 'name_to_canonical' not in globals():
    name_to_canonical = {name.lower(): name.lower() for name in person_names}
if 'invalid_pairs' not in globals():
    invalid_pairs = set()
if 'predefined_relations' not in globals():
    predefined_relations = []
if 'process_verse' not in globals():
    raise NameError("process_verse function must be defined in Cell 1")

# Gender map (expanded)
gender_map = {
    'arjuna': 'male', 'bhima': 'male', 'yudhishthira': 'male', 'nakula': 'male', 'sahadeva': 'male',
    'krishna': 'male', 'duryodhana': 'male', 'bhishma': 'male', 'karna': 'male', 'drona': 'male',
    'pandu': 'male', 'dhritarashtra': 'male', 'kunti': 'female', 'draupadi': 'female', 'subhadra': 'female',
    'gandhari': 'female', 'madri': 'female', 'devaki': 'female', 'satyavati': 'female', 'ganga': 'female',
    'ambika': 'female', 'ambalika': 'female', 'shantanu': 'male', 'vyasa': 'male', 'vidura': 'male',
    'balarama': 'male', 'drupada': 'male', 'dhrishtadyumna': 'male', 'shakuni': 'male', 'asvatthama': 'male',
    'kripa': 'male', 'sanjaya': 'male', 'abhimanyu': 'male', 'parikshit': 'male', 'ghatotkacha': 'male',
    'hidimbi': 'female', 'yaudheya': 'male', 'niramitra': 'male', 'suhotra': 'male', 'vrishasena': 'male',
    'sushena': 'male', 'iravan': 'male', 'babhruvahana': 'male', 'ulupi': 'female', 'chitrangada': 'female',
    'uttara': 'female', 'laxman': 'male', 'bhanumati': 'female', 'vrushali': 'female', 'jayadratha': 'male',
    'dushala': 'female', 'prativindhya': 'male', 'sutasoma': 'male', 'shrutasena': 'male', 'shrutakirti': 'male',
    'shrutakarma': 'male', 'satyaki': 'male', 'bhurisrava': 'male', 'somadatta': 'male', 'yodhamanyu': 'male',
    'devika': 'female', 'karenumati': 'female', 'vasudeva (father)': 'male', 'rohini': 'female',
    'jambavati': 'female', 'satyabhama': 'female', 'rukmini': 'female', 'charudeshna': 'male',
    'amba': 'female', 'vishvamitra': 'male', 'narada': 'male', 'shesha': 'male', 'kapila': 'male',
    'kashyapa': 'male', 'vivasvan': 'male', 'varuna': 'male', 'kubera': 'male', 'agni': 'male',
    'sini': 'male', 'vinda': 'male', 'alambusha': 'male', 'ugrasena': 'male', 'saurin': 'male',
    'shrutayudha': 'male', 'chitrasena': 'male', 'kekaya': 'male', 'ekalavya': 'male', 'srinjaya': 'male',
    'brihadkshatra': 'male', 'satyajit': 'male', 'sikhandi': 'female', 'rohini (mother)': 'female'
}
gender_map_lower = {k.lower(): v for k, v in gender_map.items()}
print("Sample gender_map entries:", {k: v for k, v in list(gender_map_lower.items())[:10]})

# Define relation priority for contradiction resolution
RELATION_PRIORITY = {
    'father': 1, 'mother': 1, 'son': 2, 'daughter': 2, 'husband': 3, 'wife': 3,
    'brother': 4, 'sister': 4, 'grandfather': 5, 'grandmother': 5, 'grandson': 5,
    'great-grandfather': 6, 'great-grandmother': 6, 'great-grandson': 6,
    'great-great-grandfather': 7, 'great-great-grandmother': 7, 'great-great-grandson': 7,
    'father-in-law': 8, 'mother-in-law': 8, 'son-in-law': 8, 'daughter-in-law': 8,
    'brother-in-law': 9, 'uncle': 10, 'aunt': 10, 'nephew': 10, 'niece': 10,
    'teacher': 11, 'disciple': 11, 'guardian': 12, 'ward': 12, 'friend': 13, 'ally': 14, 'cousin': 15
}

# Symmetric relations
symmetric_relations = {'brother', 'sister', 'cousin', 'friend', 'ally', 'brother-in-law'}

# Expanded invalid_pairs
invalid_pairs = invalid_pairs | {
    ('dhritarashtra', 'son', 'pandu'), ('pandu', 'son', 'drupada'), ('yodhamanyu', 'son', 'subhadra'),
    ('yodhamanyu', 'son', 'draupadi'), ('bhishma', 'son', 'somadatta'), ('krishna', 'son', 'pandu'),
    ('drupada', 'son', 'draupadi'), ('drupada', 'lord', 'subhadra'), ('drupada', 'son', 'subhadra'),
    ('brahma', 'son', 'kunti'), ('dhritarashtra', 'family', 'krishna'), ('sanjaya', 'lord', 'arjuna'),
    ('sanjaya', 'foes', 'krishna'), ('sanjaya', 'lord', 'krishna'), ('krishna', 'lord', 'arjuna'),
    ('yodhamanyu', 'son', 'vasudeva (father)'), ('uttara', 'sister', 'parikshit'),
    ('ulupi', 'mother', 'vidura'), ('arjuna', 'son', 'srinjaya'), ('bhanumati', 'mother', 'shrutasena'),
    ('shesha', 'son', 'kekaya'), ('satyavati', 'mother', 'brihadkshatra'), ('kripa', 'son', 'vasudeva (father)'),
    ('arjuna', 'great-grandson', 'bhishma'), ('dhritarashtra', 'great-great-grandson', 'satyavati'),
    ('gandhari', 'sister', 'prativindhya'), ('asvatthama', 'son', 'gandhari'),
    ('ambalika', 'mother', 'charudeshna'), ('chitrangada', 'mother', 'brihadkshatra'),
    ('devika', 'mother', 'vidura'), ('dushala', 'mother', 'ugrasena')
}

# Output directory
save_dir = "/content/drive/My Drive/Gita_Pipeline_Output"
os.makedirs(save_dir, exist_ok=True)

# Process all verses and collect relations
all_tagged = []
all_relations = []
all_untagged_keywords = []
verse_relations = []
verse_tagged = []
seen_relations = {}  # Track highest-priority relation per entity pair

for idx, row in gita_data.iterrows():
    chapter, verse, text = row['Chapter'], row['Verse'], row['Text']
    annotated, relations, untagged = process_verse(chapter, verse, text)
    verse_tagged.append((idx, annotated))
    verse_relations.append((idx, relations))
    all_untagged_keywords.extend(untagged)

# Normalize, validate, and clean relations with early contradiction resolution
all_relations_set = set()
self_loops = []
contradictions = []
valid_relations = set(RELATION_PRIORITY.keys())

for idx, relations in verse_relations:
    for rel in relations:
        e1 = name_to_canonical.get(rel[0].lower(), rel[0].lower())
        e2 = name_to_canonical.get(rel[2].lower(), rel[2].lower())
        rel_type = rel[1].lower()
        rel_norm = (e1, rel_type, e2)

        # Validate relation type
        if rel_type not in valid_relations:
            print(f"Invalid relation type skipped: {rel_norm}")
            continue

        # Validate gender constraints
        e1_gender = gender_map_lower.get(e1, 'unknown')
        e2_gender = gender_map_lower.get(e2, 'unknown')
        print(f"Gender check: {rel_norm} (e1: {e1_gender}, e2: {e2_gender})")
        if e1_gender == 'unknown' or e2_gender == 'unknown':
            print(f"Unknown gender skipped: {rel_norm}")
            continue
        if rel_type in ['mother', 'wife', 'daughter', 'sister', 'grandmother', 'great-grandmother', 'great-great-grandmother', 'mother-in-law', 'daughter-in-law', 'aunt', 'niece'] and e1_gender != 'female':
            print(f"Invalid (Gender: {e1_gender}, {e2_gender}): {rel_norm}")
            continue
        if rel_type in ['father', 'husband', 'son', 'brother', 'grandfather', 'great-grandfather', 'great-great-grandfather', 'father-in-law', 'son-in-law', 'uncle', 'nephew'] and e1_gender != 'male':
            print(f"Invalid (Gender: {e1_gender}, {e2_gender}): {rel_norm}")
            continue
        if rel_type == 'husband' and e2_gender != 'female':
            print(f"Invalid (Gender: {e1_gender}, {e2_gender}): {rel_norm}")
            continue
        if rel_type == 'wife' and e2_gender != 'male':
            print(f"Invalid (Gender: {e1_gender}, {e2_gender}): {rel_norm}")
            continue

        # Check for self-loops
        if e1 == e2:
            self_loops.append(rel_norm)
            print(f"Self-loop skipped: {rel_norm}")
            continue

        # Check for invalid pairs
        if rel_norm in invalid_pairs:
            print(f"Invalid pair skipped: {rel_norm}")
            continue

        # Check for contradictions with priority resolution
        key = (e1, e2)
        current_priority = RELATION_PRIORITY.get(rel_type, 100)
        if key in seen_relations:
            existing_rel, existing_priority = seen_relations[key]
            if existing_rel != rel_type:
                contradictions.append((e1, rel_type, e2, existing_rel))
                if current_priority < existing_priority:
                    seen_relations[key] = (rel_type, current_priority)
                    all_relations_set.discard((e1, existing_rel, e2))
                    all_relations_set.add(rel_norm)
                    print(f"Valid: {rel_norm}")
                continue
        else:
            seen_relations[key] = (rel_type, current_priority)
            all_relations_set.add(rel_norm)
            print(f"Valid: {rel_norm}")

# Log self-loops and contradictions
print(f"Found {len(self_loops)} self-loops: {self_loops[:5] or 'None'}")
print(f"Found {len(contradictions)} contradictions: {contradictions[:5]}")

# Add predefined relations with contradiction resolution
predefined_train = []
predefined_dev = []
predefined_test = []
random.seed(42)
for rel in predefined_relations:
    e1 = name_to_canonical.get(rel[0].lower(), rel[0].lower())
    e2 = name_to_canonical.get(rel[2].lower(), rel[2].lower())
    rel_type = rel[1].lower()
    rel_norm = (e1, rel_type, e2)
    if rel_norm in invalid_pairs or e1 == e2 or rel_type not in valid_relations:
        continue
    e1_gender = gender_map_lower.get(e1, 'unknown')
    e2_gender = gender_map_lower.get(e2, 'unknown')
    if (e1_gender != 'unknown' and e2_gender != 'unknown' and
        not (rel_type in ['mother', 'wife', 'daughter', 'sister', 'grandmother', 'great-grandmother', 'great-great-grandmother', 'mother-in-law', 'daughter-in-law', 'aunt', 'niece'] and e1_gender != 'female') and
        not (rel_type in ['father', 'husband', 'son', 'brother', 'grandfather', 'great-grandfather', 'great-great-grandfather', 'father-in-law', 'son-in-law', 'uncle', 'nephew'] and e1_gender != 'male') and
        not (rel_type == 'husband' and e2_gender != 'female') and
        not (rel_type == 'wife' and e2_gender != 'male')):
        key = (e1, e2)
        current_priority = RELATION_PRIORITY.get(rel_type, 100)
        if key in seen_relations:
            existing_rel, existing_priority = seen_relations[key]
            if current_priority < existing_priority:
                seen_relations[key] = (rel_type, current_priority)
                all_relations_set.discard((e1, existing_rel, e2))
                all_relations_set.add(rel_norm)
        else:
            seen_relations[key] = (rel_type, current_priority)
            all_relations_set.add(rel_norm)
        rand = random.random()
        if rand < 0.7:
            predefined_train.append(rel_norm)
        elif rand < 0.85:
            predefined_dev.append(rel_norm)
        else:
            predefined_test.append(rel_norm)

all_relations_list = list(all_relations_set)

# Generate negative samples
negative_relations = []
possible_pairs = list(combinations(person_names, 2))
random.shuffle(possible_pairs)
valid_entities = set(gender_map_lower.keys())
all_rel_types = list(RELATION_PRIORITY.keys())
random.shuffle(all_rel_types)
generated = 0
target_neg = len(all_relations_list)  # Match positive relation count
for p1, p2 in possible_pairs:
    if generated >= target_neg:
        break
    p1, p2 = p1.lower(), p2.lower()
    if p1 not in valid_entities or p2 not in valid_entities:
        continue
    for rel in all_rel_types:
        rel_tuple = (p1, rel, p2)
        e1_gender = gender_map_lower.get(p1, 'unknown')
        e2_gender = gender_map_lower.get(p2, 'unknown')
        if (rel_tuple not in all_relations_set and rel_tuple not in invalid_pairs and p1 != p2 and
            e1_gender != 'unknown' and e2_gender != 'unknown' and
            (rel not in ['mother', 'wife', 'daughter', 'sister', 'grandmother', 'great-grandmother', 'great-great-grandmother', 'mother-in-law', 'daughter-in-law', 'aunt', 'niece'] or e1_gender == 'female') and
            (rel not in ['father', 'husband', 'son', 'brother', 'grandfather', 'great-grandfather', 'great-great-grandfather', 'father-in-law', 'son-in-law', 'uncle', 'nephew'] or e1_gender == 'male') and
            (rel != 'husband' or e2_gender == 'female') and
            (rel != 'wife' or e2_gender == 'male')):
            negative_relations.append(rel_tuple)
            generated += 1
            break

print(f"Generated {len(negative_relations)} negative relations.")

# Combine positive and negative relations
all_relations_list.extend(negative_relations)

# Balance relation types
rel_counts = Counter(rel for _, rel, _ in all_relations_list)
max_samples_per_class = 15
balanced_relations = []
rel_groups = defaultdict(list)
for rel_tuple in all_relations_list:
    rel_groups[rel_tuple[1]].append(rel_tuple)
for rel, group in rel_groups.items():
    if rel not in RELATION_PRIORITY:
        print(f"Warning: Relation type '{rel}' not in priority list. Skipping balancing.")
        continue
    samples = min(len(group), max_samples_per_class)
    balanced_relations.extend(random.sample(group, samples))
    # Oversample rare classes
    if len(group) < max_samples_per_class:
        additional = min(max_samples_per_class - len(group), len(group))
        balanced_relations.extend(random.sample(group, additional))

print(f"Balanced relations: {len(balanced_relations)} from {len(all_relations_list)}")

# Clean relations
def clean_relations(relations, invalid_set, symmetric_relations):
    cleaned_relations = [r for r in relations if r not in invalid_set]
    print(f"Removed {len(relations) - len(cleaned_relations)} invalid relations.")

    seen = set()
    dedup_relations = []
    for source, relation, target in cleaned_relations:
        if relation in symmetric_relations:
            key = tuple(sorted([source, target]) + [relation])
        else:
            key = (source, relation, target)
        if key not in seen:
            seen.add(key)
            dedup_relations.append((source, relation, target))
    print(f"Deduplicated {len(cleaned_relations) - len(dedup_relations)} symmetric relations.")

    return dedup_relations

# Augment training relations
new_relations = [
    ('arjuna', 'husband', 'subhadra'), ('arjuna', 'husband', 'draupadi'), ('arjuna', 'father', 'abhimanyu'),
    ('arjuna', 'son', 'pandu'), ('arjuna', 'son', 'kunti'), ('arjuna', 'brother', 'bhima'),
    ('arjuna', 'brother', 'yudhishthira'), ('arjuna', 'brother', 'nakula'), ('arjuna', 'brother', 'sahadeva'),
    ('krishna', 'brother-in-law', 'arjuna'), ('krishna', 'ally', 'arjuna'), ('krishna', 'teacher', 'arjuna'),
    ('krishna', 'brother', 'balarama'), ('subhadra', 'sister', 'krishna'), ('subhadra', 'mother', 'abhimanyu'),
    ('kunti', 'mother', 'yudhishthira'), ('kunti', 'mother', 'bhima'), ('kunti', 'mother', 'arjuna'),
    ('pandu', 'father', 'yudhishthira'), ('pandu', 'father', 'bhima'), ('pandu', 'father', 'arjuna'),
    ('pandu', 'father', 'nakula'), ('pandu', 'father', 'sahadeva'), ('bhishma', 'grandfather', 'pandu'),
    ('bhishma', 'grandfather', 'dhritarashtra'), ('drona', 'teacher', 'arjuna'), ('drona', 'teacher', 'bhima'),
    ('drona', 'teacher', 'yudhishthira'), ('draupadi', 'wife', 'yudhishthira'), ('draupadi', 'wife', 'bhima'),
    ('draupadi', 'wife', 'nakula'), ('draupadi', 'wife', 'sahadeva'), ('yudhishthira', 'brother', 'bhima'),
    ('nakula', 'brother', 'sahadeva'), ('bhima', 'father', 'ghatotkacha'), ('gandhari', 'mother', 'duryodhana'),
    ('madri', 'mother', 'nakula'), ('madri', 'mother', 'sahadeva'), ('drupada', 'father', 'draupadi'),
    ('drupada', 'father', 'dhrishtadyumna'), ('devaki', 'mother', 'krishna'), ('devaki', 'mother', 'subhadra'),
    ('devaki', 'mother', 'balarama'), ('kunti', 'aunt', 'krishna'), ('krishna', 'nephew', 'kunti'),
    ('gandhari', 'wife', 'dhritarashtra'), ('ambika', 'mother', 'dhritarashtra'), ('ambalika', 'mother', 'pandu'),
    ('arjuna', 'grandson', 'bhishma'), ('bhima', 'grandson', 'bhishma'), ('yudhishthira', 'grandson', 'bhishma'),
    ('nakula', 'grandson', 'bhishma'), ('sahadeva', 'grandson', 'bhishma'), ('shakuni', 'uncle', 'duryodhana'),
    ('duryodhana', 'nephew', 'shakuni'), ('vidura', 'uncle', 'arjuna'), ('arjuna', 'nephew', 'vidura'),
    ('vidura', 'uncle', 'bhima'), ('vidura', 'uncle', 'yudhishthira'), ('vidura', 'uncle', 'nakula'),
    ('vidura', 'uncle', 'sahadeva'), ('vidura', 'uncle', 'duryodhana'), ('balarama', 'uncle', 'abhimanyu'),
    ('krishna', 'uncle', 'parikshit'), ('parikshit', 'grandson', 'subhadra'), ('parikshit', 'grandson', 'arjuna'),
    ('draupadi', 'daughter', 'drupada'), ('dushala', 'sister', 'duryodhana'), ('dushala', 'niece', 'shakuni'),
    ('satyavati', 'mother', 'chitrangada'), ('kunti', 'sister', 'vasudeva (father)'), ('vasudeva (father)', 'father', 'krishna'),
    ('gandhari', 'aunt', 'abhimanyu'), ('gandhari', 'aunt', 'parikshit'), ('satyavati', 'grandmother', 'pandu'),
    ('satyavati', 'grandmother', 'dhritarashtra'), ('kunti', 'aunt', 'subhadra'), ('kunti', 'aunt', 'balarama'),
    ('subhadra', 'niece', 'kunti'), ('balarama', 'nephew', 'kunti'), ('dushala', 'daughter', 'gandhari'),
    ('dushala', 'daughter', 'dhritarashtra')
]
def augment_relations(relations, new_relations, valid_entities, seen_relations):
    relation_set = set(relations)
    added = 0
    for source, rel, target in new_relations:
        source, rel, target = source.lower(), rel.lower(), target.lower()
        if (source in valid_entities and target in valid_entities and
            (source, rel, target) not in relation_set and
            rel in RELATION_PRIORITY and source != target and
            (source, rel, target) not in invalid_pairs):
            e1_gender = gender_map_lower.get(source, 'unknown')
            e2_gender = gender_map_lower.get(target, 'unknown')
            if e1_gender == 'unknown' or e2_gender == 'unknown':
                continue
            if rel in ['mother', 'wife', 'daughter', 'sister', 'grandmother', 'great-grandmother', 'great-great-grandmother', 'mother-in-law', 'daughter-in-law', 'aunt', 'niece'] and e1_gender != 'female':
                continue
            if rel in ['father', 'husband', 'son', 'brother', 'grandfather', 'great-grandfather', 'great-great-grandfather', 'father-in-law', 'son-in-law', 'uncle', 'nephew'] and e1_gender != 'male':
                continue
            if rel == 'husband' and e2_gender != 'female':
                continue
            if rel == 'wife' and e2_gender != 'male':
                continue
            key = (source, target)
            current_priority = RELATION_PRIORITY.get(rel, 100)
            if key in seen_relations:
                existing_rel, existing_priority = seen_relations[key]
                if current_priority >= existing_priority:
                    continue
                seen_relations[key] = (rel, current_priority)
                relation_set.discard((source, existing_rel, target))
            else:
                seen_relations[key] = (rel, current_priority)
            relations.append((source, rel, target))
            relation_set.add((source, rel, target))
            added += 1
    print(f"Added {added} new relations.")
    return relations

# Clean and augment
valid_entities = set(r[0] for r in balanced_relations) | set(r[2] for r in balanced_relations)
cleaned_relations = clean_relations(balanced_relations, invalid_pairs, symmetric_relations)
augmented_relations = augment_relations(cleaned_relations, new_relations, valid_entities, seen_relations)

# Resolve contradictions in augmented relations
resolved_relations = []
seen_pairs = {}
for source, rel, target in augmented_relations:
    key = (source, target)
    current_priority = RELATION_PRIORITY.get(rel, 100)
    if key in seen_pairs:
        existing_rel, existing_priority = seen_pairs[key]
        if current_priority < existing_priority:
            seen_pairs[key] = (rel, current_priority)
    else:
        seen_pairs[key] = (rel, current_priority)
resolved_relations = [(source, rel, target) for source, rel, target in augmented_relations if seen_pairs.get((source, target), [None, float('inf')])[0] == rel]
print(f"Resolved {len(augmented_relations) - len(resolved_relations)} contradictions in augmented relations.")

# Split relations into train/dev/test (70:15:15)
train_rels, temp_rels = train_test_split(resolved_relations, test_size=0.30, random_state=42)
dev_rels, test_rels = train_test_split(temp_rels, test_size=0.5, random_state=42)

# Ensure negative relations are included in splits
train_neg = []
dev_neg = []
test_neg = []
random.seed(42)
for rel in negative_relations:
    rand = random.random()
    if rand < 0.7:
        train_neg.append(rel)
    elif rand < 0.85:
        dev_neg.append(rel)
    else:
        test_neg.append(rel)
train_rels.extend(train_neg)
dev_rels.extend(dev_neg)
test_rels.extend(test_neg)

# Ensure predefined relations are included
train_rels.extend([rel for rel in predefined_train if rel not in train_rels])
dev_rels.extend([rel for rel in predefined_dev if rel not in dev_rels])
test_rels.extend([rel for rel in predefined_test if rel not in test_rels])

# Resolve contradictions in splits
def resolve_split_contradictions(relations):
    seen_pairs = {}
    resolved = []
    for source, rel, target in relations:
        key = (source, target)
        current_priority = RELATION_PRIORITY.get(rel, 100)
        if key in seen_pairs:
            existing_rel, existing_priority = seen_pairs[key]
            if current_priority < existing_priority:
                seen_pairs[key] = (rel, current_priority)
        else:
            seen_pairs[key] = (rel, current_priority)
    resolved = [(source, rel, target) for source, rel, target in relations if seen_pairs.get((source, target), [None, float('inf')])[0] == rel]
    print(f"Resolved {len(relations) - len(resolved)} contradictions in split.")
    return resolved

train_rels = resolve_split_contradictions(train_rels)
dev_rels = resolve_split_contradictions(dev_rels)
test_rels = resolve_split_contradictions(test_rels)

# Assign verses and relations to splits
train_verses = []
dev_verses = []
test_verses = []
train_relations = []
dev_relations = []
test_relations = []
verse_to_splits = defaultdict(list)

# Assign verse-extracted relations
for idx, relations in verse_relations:
    for rel in relations:
        e1 = name_to_canonical.get(rel[0].lower(), rel[0].lower())
        e2 = name_to_canonical.get(rel[2].lower(), rel[2].lower())
        rel_norm = (e1, rel[1].lower(), e2)
        if rel_norm in train_rels and rel_norm not in train_relations:
            verse_to_splits[idx].append(('train', rel_norm))
            train_relations.append(rel_norm)
        if rel_norm in dev_rels and rel_norm not in dev_relations:
            verse_to_splits[idx].append(('dev', rel_norm))
            dev_relations.append(rel_norm)
        if rel_norm in test_rels and rel_norm not in test_relations:
            verse_to_splits[idx].append(('test', rel_norm))
            test_relations.append(rel_norm)

# Add predefined and negative relations to splits
train_relations.extend([rel for rel in train_rels if rel not in train_relations])
dev_relations.extend([rel for rel in dev_rels if rel not in dev_relations])
test_relations.extend([rel for rel in test_rels if rel not in test_relations])

# Add verses to splits
for idx, splits in verse_to_splits.items():
    tagged = next(t for i, t in verse_tagged if i == idx)
    for split, rel in splits:
        if split == 'train':
            train_verses.append((idx, tagged, [rel]))
        elif split == 'dev':
            dev_verses.append((idx, tagged, [rel]))
        else:
            test_verses.append((idx, tagged, [rel]))

# Handle remaining verses
remaining_verses = [(idx, tagged) for idx, tagged in verse_tagged if idx not in verse_to_splits]
train_rem, temp_rem = train_test_split(remaining_verses, test_size=0.30, random_state=42)
dev_rem, test_rem = train_test_split(temp_rem, test_size=0.5, random_state=42)

train_verses.extend([(idx, tagged, []) for idx, tagged in train_rem])
dev_verses.extend([(idx, tagged, []) for idx, tagged in dev_rem])
test_verses.extend([(idx, tagged, []) for idx, tagged in test_rem])

train_verses = sorted(train_verses, key=lambda x: x[0])
dev_verses = sorted(dev_verses, key=lambda x: x[0])
test_verses = sorted(test_verses, key=lambda x: x[0])

# Validate relations
def validate_relations(relations, split_name):
    invalid_found = []
    self_loops_found = []
    contradictions_found = []
    seen = {}
    for source, rel, target in relations:
        if (source, rel, target) in invalid_pairs:
            invalid_found.append((source, rel, target))
        if source == target:
            self_loops_found.append((source, rel, target))
        key = (source, target)
        if key in seen and seen[key] != rel:
            contradictions_found.append((source, rel, target, seen[key]))
        seen[key] = rel
    if invalid_found:
        print(f"Warning: Found {len(invalid_found)} invalid relations in {split_name}: {invalid_found[:5]}")
    if self_loops_found:
        print(f"Warning: Found {len(self_loops_found)} self-loops in {split_name}: {self_loops_found[:5]}")
    if contradictions_found:
        print(f"Warning: Found {len(contradictions_found)} contradictions in {split_name}: {contradictions_found[:5]}")
    if len(relations) < 50 and split_name == "train":
        print(f"Warning: Low relation count in {split_name}: {len(relations)} (expected ~500)")
    if len(relations) < 10 and split_name in ["dev", "test"]:
        print(f"Warning: Low relation count in {split_name}: {len(relations)} (expected ~100)")
    return len(contradictions_found) == 0

validate_relations(train_relations, "train")
validate_relations(dev_relations, "dev")
validate_relations(test_relations, "test")

# Log class distribution
def log_class_distribution(relations, split_name):
    rel_counts = Counter(rel for _, rel, _ in relations)
    print(f"\nRelation type distribution in {split_name}:")
    for rel in sorted(RELATION_PRIORITY.keys()):
        count = rel_counts.get(rel, 0)
        print(f" - {rel}: {count}")
    return rel_counts

log_class_distribution(train_relations, "Train")
log_class_distribution(dev_relations, "Dev")
log_class_distribution(test_relations, "Test")
log_class_distribution(train_relations + dev_relations + test_relations, "All splits combined")

# Write output files
def write_tagged_file(filename, verses):
    with open(filename, 'w', encoding='utf-8') as f:
        for idx, tagged, rels in verses:
            f.write(f"# Verse {idx}\n")
            for word, tag in tagged:
                if tag not in ['O', 'B-PER', 'I-PER', 'B-LOC', 'I-LOC', 'B-REL', 'I-REL']:
                    print(f"Warning: Invalid tag '{tag}' for word '{word}' in {filename}")
                    tag = 'O'
                f.write(f"{word} {tag}\n")
            if rels:
                f.write("# Relations:\n")
                for rel in rels:
                    f.write(f"# {rel[0]} -> {rel[1]} -> {rel[2]}\n")
            f.write("\n")

def write_relations_file(filename, relations):
    seen = set()
    unique_relations = []
    for rel in relations:
        rel_tuple = (rel[0], rel[1], rel[2])
        if rel_tuple not in seen:
            seen.add(rel_tuple)
            unique_relations.append(rel)
    with open(filename, 'w', encoding='utf-8') as f:
        for subj, rel, obj in unique_relations:
            f.write(f"{subj} -> {rel} -> {obj}\n")

# Save files
write_tagged_file(os.path.join(save_dir, "train.txt"), train_verses)
write_tagged_file(os.path.join(save_dir, "dev.txt"), dev_verses)
write_tagged_file(os.path.join(save_dir, "test.txt"), test_verses)
write_relations_file(os.path.join(save_dir, "train_relations_updated.txt"), train_relations)
write_relations_file(os.path.join(save_dir, "dev_relations_updated.txt"), dev_relations)
write_relations_file(os.path.join(save_dir, "test_relations_updated.txt"), test_relations)

# Additional logging
print("\nRelation type distribution (Train):")
relation_counts = defaultdict(int)
for _, rel, _ in train_relations:
    relation_counts[rel] += 1
print(dict(relation_counts))

print("\nSample Tagged Verses (Train, first 2):")
for idx, tagged, rels in train_verses[:2]:
    print(f"Verse {idx}:")
    print("\n".join(f"{word} {tag}" for word, tag in tagged))
    if rels:
        print("Relations:")
        for rel in rels:
            print(f"  {rel[0]} -> {rel[1]} -> {rel[2]}")
    print()

print("\nSample Extracted Relations (Train, first 5 non-predefined):")
extracted = [rel for rel in train_relations if rel not in predefined_relations][:5]
for rel in extracted:
    print(f"{rel[0]} -> {rel[1]} -> {rel[2]}")

print("\nSample Negative Relations (Train, first 5):")
neg_samples = [rel for rel in train_relations if rel in negative_relations][:5]
if neg_samples:
    print("Warning: Verify negative relations for Mahabharata accuracy:")
    for rel in neg_samples:
        print(f"{rel[0]} -> {rel[1]} -> {rel[2]}")
else:
    print("Warning: No negative relations in train split. Check negative sampling logic.")

print("Files saved to Google Drive. Run the next cell to build and visualize the graph.")

# Final confirmation
print(f"\nTotal unique relations: {len(set(resolved_relations + negative_relations))} (Positive: {len(all_relations_set)}, Negative: {len(negative_relations)})")
print(f"Train relations: {len(set(train_relations))} (sample: {train_relations[:5]})")
print(f"Dev relations: {len(set(dev_relations))} (sample: {dev_relations[:5]})")
print(f"Test relations: {len(set(test_relations))} (sample: {test_relations[:5]})")
print(f"Files saved to {save_dir}")
print("Cell 2: Data processing, cleaning, and augmentation completed. Run the next cell to build and visualize the graph.")
```

```python
import random
from itertools import combinations
from collections import defaultdict, Counter
from sklearn.model_selection import train_test_split
import os

# Dependency checks
if 'gita_data' not in globals():
    raise NameError("gita_data must be defined in Cell 1")
if 'person_names' not in globals():
    raise NameError("person_names must be defined in Cell 1")
if 'name_to_canonical' not in globals():
    name_to_canonical = {name.lower(): name.lower() for name in person_names}
if 'invalid_pairs' not in globals():
    invalid_pairs = set()
if 'predefined_relations' not in globals():
    predefined_relations = [
        ('arjuna', 'husband', 'draupadi'), ('arjuna', 'husband', 'subhadra'), ('arjuna', 'husband', 'ulupi'),
        ('arjuna', 'husband', 'chitrangada'), ('kunti', 'mother', 'arjuna'), ('pandu', 'father', 'arjuna'),
        ('kunti', 'mother', 'bhima'), ('kunti', 'mother', 'yudhishthira'), ('pandu', 'father', 'bhima'),
        ('pandu', 'father', 'yudhishthira'), ('krishna', 'brother-in-law', 'arjuna'), ('bhishma', 'grandfather', 'pandu'),
        ('subhadra', 'mother', 'abhimanyu'), ('arjuna', 'father', 'abhimanyu'), ('dhritarashtra', 'brother', 'pandu'),
        ('arjuna', 'brother', 'bhima'), ('nakula', 'brother', 'sahadeva'), ('bhima', 'brother', 'yudhishthira'),
        ('drona', 'teacher', 'arjuna'), ('bhishma', 'teacher', 'arjuna'), ('krishna', 'friend', 'arjuna'),
        ('drupada', 'father', 'draupadi'), ('duryodhana', 'son', 'dhritarashtra'), ('gandhari', 'mother', 'duryodhana'),
        ('bhishma', 'teacher', 'duryodhana'), ('bhishma', 'teacher', 'bhima'), ('krishna', 'teacher', 'arjuna'),
        ('kripa', 'teacher', 'asvatthama'), ('madri', 'mother', 'nakula'), ('madri', 'mother', 'sahadeva'),
        ('pandu', 'father', 'nakula'), ('pandu', 'father', 'sahadeva'), ('devaki', 'mother', 'krishna'),
        ('vasudeva (father)', 'father', 'krishna'), ('ganga', 'mother', 'bhishma'), ('shantanu', 'father', 'bhishma'),
        ('satyavati', 'mother', 'vyasa'), ('vyasa', 'father', 'dhritarashtra'), ('vyasa', 'father', 'pandu'),
        ('vyasa', 'father', 'vidura'), ('hidimbi', 'mother', 'ghatotkacha'), ('bhima', 'father', 'ghatotkacha'),
        ('drona', 'father', 'asvatthama'), ('somadatta', 'father', 'bhurisrava'), ('bhurisrava', 'son', 'somadatta'),
        ('karna', 'friend', 'duryodhana'), ('satyaki', 'ally', 'arjuna'), ('drona', 'teacher', 'bhima'),
        ('drona', 'teacher', 'yudhishthira'), ('drona', 'teacher', 'nakula'), ('drona', 'teacher', 'sahadeva'),
        ('kripa', 'teacher', 'arjuna'), ('kripa', 'teacher', 'bhima'), ('kripa', 'teacher', 'yudhishthira'),
        ('kripa', 'teacher', 'nakula'), ('kripa', 'teacher', 'sahadeva'), ('shakuni', 'uncle', 'duryodhana'),
        ('shakuni', 'uncle', 'dushasana'), ('shakuni', 'uncle', 'vikarna'), ('vidura', 'uncle', 'arjuna'),
        ('vidura', 'uncle', 'bhima'), ('vidura', 'uncle', 'yudhishthira'), ('vidura', 'uncle', 'nakula'),
        ('vidura', 'uncle', 'sahadeva'), ('balarama', 'brother', 'krishna'), ('subhadra', 'sister', 'krishna'),
        ('dhrishtadyumna', 'brother', 'draupadi'), ('uttara', 'wife', 'abhimanyu'), ('bhanumati', 'wife', 'duryodhana'),
        ('vrushali', 'wife', 'karna'), ('gandhari', 'wife', 'dhritarashtra'), ('kunti', 'wife', 'pandu'),
        ('madri', 'wife', 'pandu'), ('satyavati', 'wife', 'shantanu'), ('ambika', 'wife', 'vichitravirya'),
        ('ambalika', 'wife', 'vichitravirya'), ('dushala', 'sister', 'duryodhana'), ('dushala', 'sister', 'dushasana'),
        ('dushala', 'sister', 'vikarna'), ('yudhishthira', 'father', 'prativindhya'), ('draupadi', 'mother', 'prativindhya'),
        ('bhima', 'father', 'sutasoma'), ('draupadi', 'mother', 'sutasoma'), ('arjuna', 'father', 'shrutakarma'),
        ('draupadi', 'mother', 'shrutakarma'), ('nakula', 'father', 'shrutasena'), ('draupadi', 'mother', 'shrutasena'),
        ('sahadeva', 'father', 'shrutakirti'), ('draupadi', 'mother', 'shrutakirti'), ('karna', 'father', 'vrishasena'),
        ('karna', 'father', 'sushena'), ('arjuna', 'father', 'iravan'), ('ulupi', 'mother', 'iravan'),
        ('arjuna', 'father', 'babhruvahana'), ('chitrangada', 'mother', 'babhruvahana'), ('duryodhana', 'father', 'laxman'),
        ('bhanumati', 'mother', 'laxman'), ('dushasana', 'father', 'chitrasena'), ('vikarna', 'father', 'chitravarma'),
        ('vidura', 'uncle', 'dushasana'), ('vidura', 'uncle', 'vikarna'), ('krishna', 'cousin', 'bhima'),
        ('krishna', 'cousin', 'sahadeva'), ('balarama', 'cousin', 'nakula'), ('balarama', 'cousin', 'sahadeva'),
        ('yudhishthira', 'friend', 'krishna'), ('bhima', 'ally', 'krishna'), ('duryodhana', 'ally', 'jayadratha'),
        ('duryodhana', 'ally', 'shalya')
    ]
if 'process_verse' not in globals():
    raise NameError("process_verse function must be defined in Cell 1")

# Gender map
gender_map = {
    'arjuna': 'male', 'bhima': 'male', 'yudhishthira': 'male', 'nakula': 'male', 'sahadeva': 'male',
    'krishna': 'male', 'duryodhana': 'male', 'bhishma': 'male', 'karna': 'male', 'drona': 'male',
    'pandu': 'male', 'dhritarashtra': 'male', 'kunti': 'female', 'draupadi': 'female', 'subhadra': 'female',
    'gandhari': 'female', 'madri': 'female', 'devaki': 'female', 'satyavati': 'female', 'ganga': 'female',
    'ambika': 'female', 'ambalika': 'female', 'shantanu': 'male', 'vyasa': 'male', 'vidura': 'male',
    'balarama': 'male', 'drupada': 'male', 'dhrishtadyumna': 'male', 'shakuni': 'male', 'asvatthama': 'male',
    'kripa': 'male', 'sanjaya': 'male', 'abhimanyu': 'male', 'parikshit': 'male', 'ghatotkacha': 'male',
    'hidimbi': 'female', 'yaudheya': 'male', 'niramitra': 'male', 'suhotra': 'male', 'vrishasena': 'male',
    'sushena': 'male', 'iravan': 'male', 'babhruvahana': 'male', 'ulupi': 'female', 'chitrangada': 'female',
    'uttara': 'female', 'laxman': 'male', 'bhanumati': 'female', 'vrushali': 'female', 'jayadratha': 'male',
    'dushala': 'female', 'prativindhya': 'male', 'sutasoma': 'male', 'shrutasena': 'male', 'shrutakirti': 'male',
    'shrutakarma': 'male', 'satyaki': 'male', 'bhurisrava': 'male', 'somadatta': 'male', 'yodhamanyu': 'male',
    'devika': 'female', 'karenumati': 'female', 'vasudeva (father)': 'male', 'rohini': 'female',
    'jambavati': 'female', 'satyabhama': 'female', 'rukmini': 'female', 'charudeshna': 'male',
    'amba': 'female', 'vishvamitra': 'male', 'narada': 'male', 'shesha': 'male', 'kapila': 'male',
    'kashyapa': 'male', 'vivasvan': 'male', 'varuna': 'male', 'kubera': 'male', 'agni': 'male',
    'sini': 'male', 'vinda': 'male', 'alambusha': 'male', 'ugrasena': 'male', 'saurin': 'male',
    'shrutayudha': 'male', 'chitrasena': 'male', 'kekaya': 'male', 'ekalavya': 'male', 'srinjaya': 'male',
    'brihadkshatra': 'male', 'satyajit': 'male', 'sikhandi': 'female', 'rohini (mother)': 'female',
    'dushasana': 'male', 'vikarna': 'male', 'vichitravirya': 'male', 'parashara': 'male',
    'virata': 'male', 'shalya': 'male', 'chitravarma': 'male'
}
gender_map_lower = {k.lower(): v for k, v in gender_map.items()}
print("Sample gender_map entries:", {k: v for k, v in list(gender_map_lower.items())[:10]})

# Relation priority
RELATION_PRIORITY = {
    'father': 1, 'mother': 1, 'son': 2, 'daughter': 2, 'husband': 3, 'wife': 3,
    'brother': 4, 'sister': 4, 'grandfather': 5, 'grandmother': 5, 'grandson': 5,
    'great-grandfather': 6, 'great-grandmother': 6, 'great-grandson': 6,
    'great-great-grandfather': 7, 'great-great-grandmother': 7, 'great-great-grandson': 7,
    'father-in-law': 8, 'mother-in-law': 8, 'son-in-law': 8, 'daughter-in-law': 8,
    'brother-in-law': 9, 'uncle': 10, 'aunt': 10, 'nephew': 10, 'niece': 10,
    'teacher': 11, 'disciple': 11, 'guardian': 12, 'ward': 12, 'friend': 13, 'ally': 14, 'cousin': 15
}
symmetric_relations = {'brother', 'sister', 'cousin', 'friend', 'ally', 'brother-in-law'}
valid_relations = set(RELATION_PRIORITY.keys())

# Expanded invalid_pairs
invalid_pairs = invalid_pairs | {
    ('dhritarashtra', 'son', 'pandu'), ('pandu', 'son', 'drupada'), ('yodhamanyu', 'son', 'subhadra'),
    ('yodhamanyu', 'son', 'draupadi'), ('bhishma', 'son', 'somadatta'), ('krishna', 'son', 'pandu'),
    ('drupada', 'son', 'draupadi'), ('drupada', 'lord', 'subhadra'), ('drupada', 'son', 'subhadra'),
    ('brahma', 'son', 'kunti'), ('dhritarashtra', 'family', 'krishna'), ('sanjaya', 'lord', 'arjuna'),
    ('sanjaya', 'foes', 'krishna'), ('sanjaya', 'lord', 'krishna'), ('krishna', 'lord', 'arjuna'),
    ('yodhamanyu', 'son', 'vasudeva (father)'), ('uttara', 'sister', 'parikshit'),
    ('ulupi', 'mother', 'vidura'), ('arjuna', 'son', 'srinjaya'), ('bhanumati', 'mother', 'shrutasena'),
    ('shesha', 'son', 'kekaya'), ('satyavati', 'mother', 'brihadkshatra'), ('kripa', 'son', 'vasudeva (father)'),
    ('arjuna', 'great-grandson', 'bhishma'), ('dhritarashtra', 'great-great-grandson', 'satyavati'),
    ('gandhari', 'sister', 'prativindhya'), ('asvatthama', 'son', 'gandhari'),
    ('ambalika', 'mother', 'charudeshna'), ('chitrangada', 'mother', 'brihadkshatra'),
    ('devika', 'mother', 'vidura'), ('dushala', 'mother', 'ugrasena'),
    ('vikarna', 'brother', 'nakula'), ('vikarna', 'disciple', 'bhima'), ('vikarna', 'great-grandson', 'yudhishthira'),
    ('vikarna', 'uncle', 'sahadeva'), ('vikarna', 'father', 'arjuna'), ('vikarna', 'guardian', 'dushasana'),
    ('vikarna', 'ally', 'nakula'), ('drona', 'great-grandson', 'kripa'), ('drona', 'great-grandson', 'shantanu'),
    ('shantanu', 'friend', 'drona'), ('shantanu', 'son-in-law', 'kripa'), ('subhadra', 'wife', 'krishna'),
    ('subhadra', 'wife', 'balarama'), ('bhima', 'great-grandfather', 'vikarna'), ('bhima', 'great-grandfather', 'dushasana'),
    ('nakula', 'brother-in-law', 'dushasana'), ('nakula', 'brother-in-law', 'duryodhana'), ('kripa', 'uncle', 'bhishma'),
    ('kripa', 'son-in-law', 'vyasa'), ('bhima', 'disciple', 'sahadeva'), ('krishna', 'ward', 'satyaki'),
    ('subhadra', 'disciple', 'satyaki'), ('shantanu', 'ward', 'vyasa'), ('bhima', 'uncle', 'duryodhana')
}

# Expanded known_correct_relations
known_correct_relations = {
    ('arjuna', 'husband', 'draupadi'), ('arjuna', 'husband', 'subhadra'), ('arjuna', 'husband', 'ulupi'),
    ('arjuna', 'husband', 'chitrangada'), ('kunti', 'mother', 'arjuna'), ('pandu', 'father', 'arjuna'),
    ('kunti', 'mother', 'bhima'), ('kunti', 'mother', 'yudhishthira'), ('pandu', 'father', 'bhima'),
    ('pandu', 'father', 'yudhishthira'), ('krishna', 'brother-in-law', 'arjuna'), ('bhishma', 'grandfather', 'pandu'),
    ('subhadra', 'mother', 'abhimanyu'), ('arjuna', 'father', 'abhimanyu'), ('dhritarashtra', 'brother', 'pandu'),
    ('arjuna', 'brother', 'bhima'), ('nakula', 'brother', 'sahadeva'), ('bhima', 'brother', 'yudhishthira'),
    ('drona', 'teacher', 'arjuna'), ('bhishma', 'teacher', 'arjuna'), ('krishna', 'friend', 'arjuna'),
    ('drupada', 'father', 'draupadi'), ('duryodhana', 'son', 'dhritarashtra'), ('gandhari', 'mother', 'duryodhana'),
    ('bhishma', 'teacher', 'duryodhana'), ('bhishma', 'teacher', 'bhima'), ('krishna', 'teacher', 'arjuna'),
    ('kripa', 'teacher', 'asvatthama'), ('madri', 'mother', 'nakula'), ('madri', 'mother', 'sahadeva'),
    ('pandu', 'father', 'nakula'), ('pandu', 'father', 'sahadeva'), ('devaki', 'mother', 'krishna'),
    ('vasudeva (father)', 'father', 'krishna'), ('ganga', 'mother', 'bhishma'), ('shantanu', 'father', 'bhishma'),
    ('satyavati', 'mother', 'vyasa'), ('vyasa', 'father', 'dhritarashtra'), ('vyasa', 'father', 'pandu'),
    ('vyasa', 'father', 'vidura'), ('hidimbi', 'mother', 'ghatotkacha'), ('bhima', 'father', 'ghatotkacha'),
    ('drona', 'father', 'asvatthama'), ('somadatta', 'father', 'bhurisrava'), ('bhurisrava', 'son', 'somadatta'),
    ('karna', 'friend', 'duryodhana'), ('satyaki', 'ally', 'arjuna'), ('sanjaya', 'teacher', 'drona'),
    ('drona', 'disciple', 'sanjaya'), ('arjuna', 'disciple', 'krishna'), ('arjuna', 'disciple', 'bhishma'),
    ('arjuna', 'disciple', 'drona'), ('bhima', 'disciple', 'drona'), ('yudhishthira', 'disciple', 'drona'),
    ('nakula', 'disciple', 'drona'), ('sahadeva', 'disciple', 'drona'), ('duryodhana', 'disciple', 'drona'),
    ('dushasana', 'disciple', 'drona'), ('arjuna', 'disciple', 'kripa'), ('bhima', 'disciple', 'kripa'),
    ('yudhishthira', 'disciple', 'kripa'), ('nakula', 'disciple', 'kripa'), ('sahadeva', 'disciple', 'kripa'),
    ('duryodhana', 'disciple', 'kripa'), ('bhima', 'disciple', 'bhishma'), ('nakula', 'disciple', 'bhishma'),
    ('sahadeva', 'disciple', 'bhishma'), ('duryodhana', 'disciple', 'bhishma')
}

# Output directory
save_dir = "/content/drive/My Drive/Gita_Pipeline_Output"
os.makedirs(save_dir, exist_ok=True)

# Helper functions
def validate_gender(rel_tuple, gender_map_lower):
    source, rel, target = rel_tuple
    e1_gender = gender_map_lower.get(source, 'unknown')
    e2_gender = gender_map_lower.get(target, 'unknown')

    if e1_gender == 'unknown' or e2_gender == 'unknown':
        print(f"Unknown gender skipped: {rel_tuple}")
        return False

    if rel in ['mother', 'wife', 'daughter', 'sister', 'grandmother', 'great-grandmother', 'great-great-grandmother', 'mother-in-law', 'daughter-in-law', 'aunt', 'niece'] and e1_gender != 'female':
        print(f"Invalid gender (e1: {e1_gender}): {rel_tuple}")
        return False
    if rel in ['father', 'husband', 'son', 'brother', 'grandfather', 'great-grandfather', 'great-great-grandfather', 'father-in-law', 'son-in-law', 'uncle', 'nephew'] and e1_gender != 'male':
        print(f"Invalid gender (e1: {e1_gender}): {rel_tuple}")
        return False
    if rel == 'husband' and e2_gender != 'female':
        print(f"Invalid gender (e2: {e2_gender}): {rel_tuple}")
        return False
    if rel == 'wife' and e2_gender != 'male':
        print(f"Invalid gender (e2: {e2_gender}): {rel_tuple}")
        return False

    return True

def resolve_contradictions(relations, relation_priority):
    seen_pairs = {}
    resolved = []
    contradictions = []

    for source, rel, target in relations:
        key = (source, target)
        priority = relation_priority.get(rel, 100)

        if key in seen_pairs:
            existing_rel, existing_priority = seen_pairs[key]
            if existing_rel != rel:
                contradictions.append((source, rel, target, existing_rel))
                if priority < existing_priority:
                    seen_pairs[key] = (rel, priority)
        else:
            seen_pairs[key] = (rel, priority)

    for (source, target), (rel, _) in seen_pairs.items():
        resolved.append((source, rel, target))

    print(f"Resolved {len(contradictions)} contradictions: {contradictions[:5] or 'None'}")
    return resolved

def clean_relations(relations, invalid_set, symmetric_relations):
    cleaned = [r for r in relations if r not in invalid_set]
    print(f"Removed {len(relations) - len(cleaned)} invalid relations.")

    seen = set()
    deduped = []
    for source, rel, target in cleaned:
        key = tuple(sorted([source, target]) + [rel]) if rel in symmetric_relations else (source, rel, target)
        if key not in seen:
            seen.add(key)
            deduped.append((source, rel, target))

    print(f"Deduplicated {len(cleaned) - len(deduped)} symmetric relations.")
    return deduped

def augment_relations(relations, new_relations, valid_entities, seen_relations, gender_map_lower, invalid_pairs):
    relation_set = set(relations)
    added = []

    if not new_relations:
        new_relations = predefined_relations

    for source, rel, target in new_relations:
        source, rel, target = source.lower(), rel.lower(), target.lower()
        rel_tuple = (source, rel, target)

        if (source in valid_entities and target in valid_entities and
            rel_tuple not in relation_set and rel in RELATION_PRIORITY and
            source != target and rel_tuple not in invalid_pairs and
            validate_gender(rel_tuple, gender_map_lower) and
            rel_tuple in known_correct_relations):
            key = (source, target)
            current_priority = RELATION_PRIORITY.get(rel, 100)
            if key in seen_relations:
                existing_rel, existing_priority = seen_relations[key]
                if current_priority >= existing_priority:
                    continue
                relation_set.discard((source, existing_rel, target))
                seen_relations[key] = (rel, current_priority)
            else:
                seen_relations[key] = (rel, current_priority)
            relations.append(rel_tuple)
            relation_set.add(rel_tuple)
            added.append(rel_tuple)

    print(f"Added {len(added)} new relations: {added[:5]}")
    return relations

def balance_relations(relations, max_per_class=200, min_per_class=15):
    rel_counts = Counter(rel for _, rel, _ in relations)
    balanced = []

    rel_groups = defaultdict(list)
    for rel in relations:
        rel_groups[rel[1]].append(rel)

    for rel_type, group in rel_groups.items():
        count = len(group)
        if count > max_per_class:
            balanced.extend(random.sample(group, max_per_class))
        elif count < min_per_class and count > 0:
            balanced.extend(group)
            additional = random.choices(group, k=min_per_class - count)
            balanced.extend(additional)
        else:
            balanced.extend(group)

    print(f"Balanced to {len(balanced)} relations (from {len(relations)})")
    print("Relation distribution:", Counter(rel for _, rel, _ in balanced))
    return balanced

def generate_negative_relations(positive_relations, valid_entities, relation_types, target_count, invalid_pairs):
    negative_relations = []
    valid_pairs = set((r[0], r[2]) for r in positive_relations) | set((r[2], r[0]) for r in positive_relations if r[1] in symmetric_relations)
    generated = 0

    known_false_relations = {
        ('ghatotkacha', 'cousin', 'chitrasena'), ('bhima', 'great-grandfather', 'balarama'),
        ('vikarna', 'father', 'arjuna')
    }

    pandavas = {'arjuna', 'bhima', 'yudhishthira', 'nakula', 'sahadeva'}
    kauravas = {'duryodhana', 'dushasana', 'vikarna'}
    yadavas = {'krishna', 'balarama', 'satyaki', 'subhadra'}
    elders = {'bhishma', 'drona', 'kripa', 'shantanu', 'vyasa'}
    contemporaries = pandavas | kauravas | yadavas

    random.shuffle(list(valid_entities))
    for p1 in valid_entities:
        for p2 in valid_entities:
            if p1 == p2 or generated >= target_count:
                continue
            if ((p1 in pandavas and p2 in kauravas) or (p1 in kauravas and p2 in pandavas) or
                (p1 in yadavas and p2 in yadavas) or (p1 in pandavas and p2 in pandavas) or
                (p1 in kauravas and p2 in kauravas) or (p1 in elders and p2 in elders)):
                for rel in random.sample(relation_types, len(relation_types)):
                    rel_tuple = (p1, rel, p2)
                    if (rel_tuple not in positive_relations and
                        rel_tuple not in invalid_pairs and
                        (p1, p2) not in valid_pairs and
                        rel_tuple not in known_false_relations):
                        e1_gender = gender_map_lower.get(p1, 'unknown')
                        e2_gender = gender_map_lower.get(p2, 'unknown')
                        if (e1_gender != 'unknown' and e2_gender != 'unknown' and
                            (rel not in ['mother', 'wife', 'daughter', 'sister', 'grandmother', 'great-grandmother', 'great-great-grandmother', 'mother-in-law', 'daughter-in-law', 'aunt', 'niece'] or e1_gender == 'female') and
                            (rel not in ['father', 'husband', 'son', 'brother', 'grandfather', 'great-grandfather', 'great-great-grandfather', 'father-in-law', 'son-in-law', 'uncle', 'nephew'] or e1_gender == 'male') and
                            (rel != 'husband' or e2_gender == 'female') and
                            (rel != 'wife' or e2_gender == 'male') and
                            (rel not in ['father', 'mother', 'son', 'daughter', 'grandfather', 'grandmother', 'great-grandfather', 'great-grandmother', 'great-grandson'] or
                             (p1 in elders and p2 in contemporaries) or (p2 in elders and p1 in contemporaries))):
                            negative_relations.append(rel_tuple)
                            generated += 1
                            break
            if generated >= target_count:
                break

    print(f"Generated {len(negative_relations)} negative relations: {negative_relations[:5]}")
    return negative_relations

def validate_domain_relations(relations, known_correct):
    incorrect = []
    for rel in relations:
        if rel not in known_correct and rel not in negative_relations:
            incorrect.append(rel)

    if incorrect:
        print(f"Warning: Found {len(incorrect)} potentially incorrect relations: {incorrect[:5]}")
        print("Consider manual review or adding to known_correct_relations.")

    return relations

def validate_and_log_split(relations, split_name, expected_count):
    rel_counts = Counter(rel for _, rel, _ in relations)
    invalid_found = [r for r in relations if r in invalid_pairs]
    self_loops = [r for r in relations if r[0] == r[2]]
    contradictions = []
    seen = {}
    for source, rel, target in relations:
        key = (source, target)
        if key in seen and seen[key] != rel:
            contradictions.append((source, rel, target, seen[key]))
        seen[key] = rel

    print(f"\nValidation for {split_name}:")
    print(f" - Total relations: {len(relations)} (expected ~{expected_count})")
    print(f" - Invalid relations: {len(invalid_found)} {invalid_found[:5] or 'None'}")
    print(f" - Self-loops: {len(self_loops)} {self_loops[:5] or 'None'}")
    print(f" - Contradictions: {len(contradictions)} {contradictions[:5] or 'None'}")
    print(f" - Relation distribution: {dict(rel_counts)}")

    return len(contradictions) == 0

# Process verses
all_relations_set = set()
self_loops = []
verse_relations = []
verse_tagged = []
validated_cache = set()

for idx, row in gita_data.iterrows():
    chapter, verse, text = row['Chapter'], row['Verse'], row['Text']
    annotated, relations, untagged = process_verse(chapter, verse, text)
    verse_tagged.append((idx, annotated))
    verse_relations.append((idx, relations))

# Normalize and validate relations
for idx, relations in verse_relations:
    for rel in relations:
        e1 = name_to_canonical.get(rel[0].lower(), rel[0].lower())
        e2 = name_to_canonical.get(rel[2].lower(), rel[2].lower())
        rel_type = rel[1].lower()
        rel_norm = (e1, rel_type, e2)

        if rel_norm in validated_cache:
            all_relations_set.add(rel_norm)
            continue

        if rel_type not in valid_relations:
            print(f"Invalid relation type skipped: {rel_norm}")
            continue

        if e1 == e2:
            self_loops.append(rel_norm)
            print(f"Self-loop skipped: {rel_norm}")
            continue

        if rel_norm in invalid_pairs:
            print(f"Invalid pair skipped: {rel_norm}")
            continue

        if validate_gender(rel_norm, gender_map_lower):
            validated_cache.add(rel_norm)
            all_relations_set.add(rel_norm)
            print(f"Valid: {rel_norm}")

print(f"Found {len(self_loops)} self-loops: {self_loops[:5] or 'None'}")

# Add predefined relations
predefined_train = []
predefined_dev = []
predefined_test = []
random.seed(42)
for rel in predefined_relations:
    e1 = name_to_canonical.get(rel[0].lower(), rel[0].lower())
    e2 = name_to_canonical.get(rel[2].lower(), rel[2].lower())
    rel_type = rel[1].lower()
    rel_norm = (e1, rel_type, e2)
    if (rel_norm not in invalid_pairs and e1 != e2 and rel_type in valid_relations and
        validate_gender(rel_norm, gender_map_lower)):
        rand = random.random()
        if rand < 0.7:
            predefined_train.append(rel_norm)
        elif rand < 0.85:
            predefined_dev.append(rel_norm)
        else:
            predefined_test.append(rel_norm)

all_relations_list = list(all_relations_set)

# Generate negative relations
valid_entities = set(gender_map_lower.keys())
all_rel_types = list(RELATION_PRIORITY.keys())
negative_relations = generate_negative_relations(all_relations_set, valid_entities, all_rel_types, len(all_relations_set), invalid_pairs)
all_relations_list.extend(negative_relations)

# Balance and clean relations
balanced_relations = balance_relations(all_relations_list, max_per_class=200, min_per_class=15)
cleaned_relations = clean_relations(balanced_relations, invalid_pairs, symmetric_relations)

# Augment relations
valid_entities = set(r[0] for r in cleaned_relations) | set(r[2] for r in cleaned_relations)
seen_relations = {(r[0], r[2]): (r[1], RELATION_PRIORITY.get(r[1], 100)) for r in cleaned_relations}
augmented_relations = augment_relations(cleaned_relations, predefined_relations, valid_entities, seen_relations, gender_map_lower, invalid_pairs)
augmented_relations = validate_domain_relations(augmented_relations, known_correct_relations)

# Resolve contradictions
resolved_relations = resolve_contradictions(augmented_relations, RELATION_PRIORITY)

# Split relations
train_rels, temp_rels = train_test_split(resolved_relations, test_size=0.30, random_state=42)
dev_rels, test_rels = train_test_split(temp_rels, test_size=0.5, random_state=42)

# Ensure unique relations per split
train_set = set(train_rels)
dev_set = set(dev_rels) - train_set
test_set = set(test_rels) - train_set - dev_set
train_rels = list(train_set)
dev_rels = list(dev_set)
test_rels = list(test_set)

# Add negative and predefined relations
train_neg = [rel for rel in negative_relations if random.random() < 0.7]
dev_neg = [rel for rel in negative_relations if 0.7 <= random.random() < 0.85]
test_neg = [rel for rel in negative_relations if random.random() >= 0.85]
train_rels.extend(train_neg)
dev_rels.extend(dev_neg)
test_rels.extend(test_neg)
train_rels.extend([rel for rel in predefined_train if rel not in train_rels])
dev_rels.extend([rel for rel in predefined_dev if rel not in dev_rels])
test_rels.extend([rel for rel in predefined_test if rel not in test_rels])

# Final contradiction resolution
train_rels = resolve_contradictions(train_rels, RELATION_PRIORITY)
dev_rels = resolve_contradictions(dev_rels, RELATION_PRIORITY)
test_rels = resolve_contradictions(test_rels, RELATION_PRIORITY)

# Assign verses and relations to splits
train_verses = []
dev_verses = []
test_verses = []
train_relations = []
dev_relations = []
test_relations = []
verse_to_splits = defaultdict(list)

for idx, relations in verse_relations:
    for rel in relations:
        e1 = name_to_canonical.get(rel[0].lower(), rel[0].lower())
        e2 = name_to_canonical.get(rel[2].lower(), rel[2].lower())
        rel_norm = (e1, rel[1].lower(), e2)
        if rel_norm in train_rels and rel_norm not in train_relations:
            verse_to_splits[idx].append(('train', rel_norm))
            train_relations.append(rel_norm)
        if rel_norm in dev_rels and rel_norm not in dev_relations:
            verse_to_splits[idx].append(('dev', rel_norm))
            dev_relations.append(rel_norm)
        if rel_norm in test_rels and rel_norm not in test_relations:
            verse_to_splits[idx].append(('test', rel_norm))
            test_relations.append(rel_norm)

train_relations.extend([rel for rel in train_rels if rel not in train_relations])
dev_relations.extend([rel for rel in dev_rels if rel not in dev_relations])
test_relations.extend([rel for rel in test_rels if rel not in test_relations])

for idx, splits in verse_to_splits.items():
    tagged = next(t for i, t in verse_tagged if i == idx)
    for split, rel in splits:
        if split == 'train':
            train_verses.append((idx, tagged, [rel]))
        elif split == 'dev':
            dev_verses.append((idx, tagged, [rel]))
        else:
            test_verses.append((idx, tagged, [rel]))

remaining_verses = [(idx, tagged) for idx, tagged in verse_tagged if idx not in verse_to_splits]
train_rem, temp_rem = train_test_split(remaining_verses, test_size=0.30, random_state=42)
dev_rem, test_rem = train_test_split(temp_rem, test_size=0.5, random_state=42)

train_verses.extend([(idx, tagged, []) for idx, tagged in train_rem])
dev_verses.extend([(idx, tagged, []) for idx, tagged in dev_rem])
test_verses.extend([(idx, tagged, []) for idx, tagged in test_rem])

train_verses = sorted(train_verses, key=lambda x: x[0])
dev_verses = sorted(dev_verses, key=lambda x: x[0])
test_verses = sorted(test_verses, key=lambda x: x[0])

# Validate and log splits
validate_and_log_split(train_relations, "train", 500)
validate_and_log_split(dev_relations, "dev", 100)
validate_and_log_split(test_relations, "test", 100)

# Write output files
def write_tagged_file(filename, verses):
    with open(filename, 'w', encoding='utf-8') as f:
        for idx, tagged, rels in verses:
            f.write(f"# Verse {idx}\n")
            for word, tag in tagged:
                if tag not in ['O', 'B-PER', 'I-PER', 'B-LOC', 'I-LOC', 'B-REL', 'I-REL']:
                    print(f"Warning: Invalid tag '{tag}' for word '{word}' in {filename}")
                    tag = 'O'
                f.write(f"{word} {tag}\n")
            if rels:
                f.write("# Relations:\n")
                for rel in rels:
                    f.write(f"# {rel[0]} -> {rel[1]} -> {rel[2]}\n")
            f.write("\n")

def write_relations_file(filename, relations):
    seen = set()
    unique_relations = []
    for rel in relations:
        rel_tuple = (rel[0], rel[1], rel[2])
        if rel_tuple not in seen:
            seen.add(rel_tuple)
            unique_relations.append(rel)
    with open(filename, 'w', encoding='utf-8') as f:
        for subj, rel, obj in unique_relations:
            f.write(f"{subj} -> {rel} -> {obj}\n")

write_tagged_file(os.path.join(save_dir, "train.txt"), train_verses)
write_tagged_file(os.path.join(save_dir, "dev.txt"), dev_verses)
write_tagged_file(os.path.join(save_dir, "test.txt"), test_verses)
write_relations_file(os.path.join(save_dir, "train_relations_updated.txt"), train_relations)
write_relations_file(os.path.join(save_dir, "dev_relations_updated.txt"), dev_relations)
write_relations_file(os.path.join(save_dir, "test_relations_updated.txt"), test_relations)

# Final logging
print("\nRelation Sources Summary:")
print(f" - Verse-extracted: {len(all_relations_set)}")
print(f" - Predefined: {len(predefined_train) + len(predefined_dev) + len(predefined_test)}")
print(f" - Negative: {len(negative_relations)}")
print(f" - Augmented: {len(augmented_relations) - len(cleaned_relations)}")
print(f"\nTotal unique relations: {len(set(resolved_relations + negative_relations))}")
print(f"Train relations: {len(set(train_relations))} (sample: {train_relations[:5]})")
print(f"Dev relations: {len(set(dev_relations))} (sample: {dev_relations[:5]})")
print(f"Test relations: {len(set(test_relations))} (sample: {test_relations[:5]})")
print(f"Files saved to {save_dir}")
print("Cell 2: Data processing, cleaning, and augmentation completed.")
```

```python
# Cell 2.5: Clean relations by removing invalid pairs and deduplicating symmetric relations

import os
from google.colab import drive

# Mount Google Drive
drive.mount('/content/drive')

# Paths to relation files
data_dir = '/content/drive/My Drive/Gita_Pipeline_Output/'
train_file = os.path.join(data_dir, 'train_relations.txt')
dev_file = os.path.join(data_dir, 'dev_relations.txt')
test_file = os.path.join(data_dir, 'test_relations.txt')
output_dir = data_dir

# Symmetric relations to deduplicate (defined here as they are specific to cleaning)
symmetric_relations = {'brother', 'friend', 'cousin', 'brother-in-law', 'ally'}
new_relations = []
def load_relations(file_path):
    """Load relations from a file into a list of tuples (source, relation, target)."""
    relations = []
    if os.path.exists(file_path):
        with open(file_path, 'r') as f:
            for line in f:
                line = line.strip()
                if line:
                    source, relation, target = map(str.strip, line.split('->'))
                    relations.append((source.lower(), relation.lower(), target.lower()))
    return relations

def save_relations(relations, file_path):
    """Save relations to a file."""
    with open(file_path, 'w') as f:
        for source, relation, target in relations:
            f.write(f"{source} -> {relation} -> {target}\n")

def clean_relations(relations, invalid_set, symmetric_relations):
    """Clean relations by removing invalids and deduplicating symmetric relations."""
    # Remove invalid relations
    cleaned_relations = [r for r in relations if r not in invalid_set]
    print(f"Removed {len(relations) - len(cleaned_relations)} invalid relations.")

    # Deduplicate symmetric relations
    seen = set()
    dedup_relations = []
    for source, relation, target in cleaned_relations:
        if relation in symmetric_relations:
            key = tuple(sorted([source, target]) + [relation])
            if key not in seen:
                seen.add(key)
                dedup_relations.append((source, relation, target))
        else:
            dedup_relations.append((source, relation, target))
    print(f"Removed {len(cleaned_relations) - len(dedup_relations)} duplicate symmetric relations.")

    return dedup_relations

def augment_relations(relations, new_relations, valid_entities):
    """Augment relations with new ones, avoiding duplicates and invalid entities."""
    relation_set = set(relations)
    added = 0
    for source, rel, target in new_relations:
        source, rel, target = source.lower(), rel.lower(), target.lower()
        if source in valid_entities and target in valid_entities and (source, rel, target) not in relation_set:
            relations.append((source, rel, target))
            relation_set.add((source, rel, target))
            added += 1
    print(f"Added {added} new relations.")
    return relations

def process_file(input_path, output_path, invalid_set, symmetric_relations, new_relations, split_name):
    """Process a single relation file."""
    print(f"\nProcessing {split_name} relations from {input_path}")
    relations = load_relations(input_path)
    print(f"Original relations: {len(relations)}")

    # Get valid entities
    valid_entities = set(source for source, _, _ in relations) | set(target for _, _, target in relations)

    # Clean relations
    cleaned_relations = clean_relations(relations, invalid_set, symmetric_relations)

    # Augment with new relations (only for train)
    if split_name == 'train':
        cleaned_relations = augment_relations(cleaned_relations, new_relations, valid_entities)

    # Save cleaned relations
    save_relations(cleaned_relations, output_path)
    print(f"Saved {len(cleaned_relations)} relations to {output_path}")

# Process all files
process_file(train_file, os.path.join(output_dir, 'train_relations_cleaned.txt'),
             invalid_pairs, symmetric_relations, new_relations, 'train')
process_file(dev_file, os.path.join(output_dir, 'dev_relations_cleaned.txt'),
             invalid_pairs, symmetric_relations, [], 'dev')
process_file(test_file, os.path.join(output_dir, 'test_relations_cleaned.txt'),
             invalid_pairs, symmetric_relations, [], 'test')

print("\nData cleaning complete. Cleaned files saved in", output_dir)
```

```python
# Cell 3: Build and Visualize the Graph (Updated with Fixes)
# This cell constructs a directed graph from the cleaned relations, validates the edges,
# and visualizes the graph for key characters in the Mahabharata.

import networkx as nx
import pickle
from pyvis.network import Network
import os
from collections import Counter

# Step 1: Load relations from cleaned files
def load_relations_from_file(filename):
    """Load relations from a file, with error handling for missing or malformed files."""
    relations = []
    try:
        if not os.path.exists(filename):
            raise FileNotFoundError(f"Relation file not found: {filename}")
        with open(filename, 'r', encoding='utf-8') as f:
            for line in f:
                line = line.strip()
                if line:
                    parts = line.split(" -> ")
                    if len(parts) != 3:
                        print(f"Warning: Malformed line in {filename}: {line}")
                        continue
                    person1, rel, person2 = parts
                    person1 = person1.lower()
                    person2 = person2.lower()
                    person1 = name_to_canonical.get(person1, person1)
                    person2 = name_to_canonical.get(person2, person2)
                    relations.append((person1, rel, person2))
    except Exception as e:
        print(f"Error loading relations from {filename}: {e}")
        return []
    return relations

# Define the directory where cleaned relation files are stored
save_dir = "/content/drive/My Drive/Gita_Pipeline_Output"

# Load relations from train, dev, and test splits
train_relations = load_relations_from_file(os.path.join(save_dir, "train_relations_updated.txt"))
dev_relations = load_relations_from_file(os.path.join(save_dir, "dev_relations_updated.txt"))
test_relations = load_relations_from_file(os.path.join(save_dir, "test_relations_updated.txt"))

# Combine all relations and deduplicate
all_relations = train_relations + dev_relations + test_relations
all_relations = list(set(all_relations))  # Deduplicate relations

# Log the number of relations loaded from each split
print("Loaded relations:")
print(f" - Train: {len(train_relations)} relations")
print(f" - Dev: {len(dev_relations)} relations")
print(f" - Test: {len(test_relations)} relations")
print(f" - Total (after deduplication): {len(all_relations)} relations")

# Step 2: Construct the graph with validation
G = nx.DiGraph()

# Add nodes with gender attribute for all valid characters
for char in person_names:
    if char in person_names_set:  # Ensure only valid persons are nodes
        G.add_node(char, gender=gender_map.get(char, "unknown"))

# Define invalid entities that should not be part of the graph
invalid_entities = {
    'mrityu', 'death', 'god', 'lord', 'divinity', 'heaven', 'soul', 'spirit', 'war', 'battle', 'chariot', 'divine',
    'fate', 'destiny', 'earth', 'sky', 'fire', 'wind', 'water', 'time', 'veda', 'scripture', 'knowledge', 'unknown',
    'brahma', 'hanuman', 'garuda', 'vasuki', 'shesha', 'vishvamitra', 'markandeya', 'bhrigu', 'vashishtha', 'chandra',
    'kapila', 'varuna', 'yama', 'kubera', 'aditya', 'narada', 'usanas', 'vivasvan', 'indra', 'shiva', 'surya', 'vayu',
    'agni', 'mahadeva', 'rudra', 'maruti', 'anjaneya'
}

# Define a knowledge base for factual validation (key relations to enforce)
known_relations = {
    ("vasudeva", "father", "krishna"): True,
    ("devaki", "mother", "krishna"): True,
    ("krishna", "brother", "balarama"): True,
    ("balarama", "brother", "krishna"): True,
    ("krishna", "sister", "subhadra"): True,
    ("subhadra", "brother", "krishna"): True,
    ("pandu", "father", "arjuna"): True,
    ("pandu", "father", "bhima"): True,
    ("pandu", "father", "yudhishthira"): True,
    ("pandu", "father", "nakula"): True,
    ("pandu", "father", "sahadeva"): True,
    ("kunti", "mother", "arjuna"): True,
    ("kunti", "mother", "bhima"): True,
    ("kunti", "mother", "yudhishthira"): True,
    ("kunti", "mother", "karna"): True,
    ("madri", "mother", "nakula"): True,
    ("madri", "mother", "sahadeva"): True,
    ("dhritarashtra", "father", "duryodhana"): True,
    ("gandhari", "mother", "duryodhana"): True,
    ("drupada", "father", "draupadi"): True,
    ("bhima", "father", "ghatotkacha"): True,
    ("hidimbi", "mother", "ghatotkacha"): True,
}

# Dictionaries to track which split each relation belongs to
relation_to_split = {}
for rel in train_relations:
    relation_to_split[rel] = "train"
for rel in dev_relations:
    relation_to_split[rel] = "dev"
for rel in test_relations:
    relation_to_split[rel] = "test"

# Lists to store updated relations (including inferred ones)
updated_train_relations = set(train_relations)
updated_dev_relations = set(dev_relations)
updated_test_relations = set(test_relations)

# Define reverse relations
reverse_relations = {
    "son": "father",
    "daughter": "mother",
    "father": "son",
    "mother": "daughter",
    "husband": "wife",
    "wife": "husband",
    "brother": "brother",
    "sister": "sister",
    "cousin": "cousin",
    "brother-in-law": "brother-in-law"
}

# Add relations to the graph with strict validation
invalid_edges_pre_add = []
for person1, rel, person2 in all_relations:
    # Skip self-loops
    if person1 == person2:
        print(f"Warning: Self-loop detected and skipped: {person1} -> {rel} -> {person2}")
        continue
    # Check for invalid entities
    if (person1 in invalid_entities or person2 in invalid_entities):
        invalid_edges_pre_add.append((person1, rel, person2))
        continue
    # Check invalid pairs
    if (person1, rel, person2) in invalid_pairs:
        invalid_edges_pre_add.append((person1, rel, person2))
        continue
    # Check if persons are valid
    if not (person1 in person_names and person2 in person_names):
        invalid_edges_pre_add.append((person1, rel, person2))
        continue
    # Gender-based validation
    p1_gender = gender_map.get(person1, 'unknown')
    p2_gender = gender_map.get(person2, 'unknown')
    if rel == 'son' and p1_gender != 'male':
        invalid_edges_pre_add.append((person1, rel, person2))
        continue
    elif rel == 'daughter' and p1_gender != 'female':
        invalid_edges_pre_add.append((person1, rel, person2))
        continue
    elif rel == 'father' and p1_gender != 'male':
        invalid_edges_pre_add.append((person1, rel, person2))
        continue
    elif rel == 'mother' and p1_gender != 'female':
        invalid_edges_pre_add.append((person1, rel, person2))
        continue
    elif rel == 'brother' and (p1_gender != 'male' or p2_gender != 'male'):
        invalid_edges_pre_add.append((person1, rel, person2))
        continue
    elif rel == 'sister' and (p1_gender != 'female' or p2_gender != 'female'):
        invalid_edges_pre_add.append((person1, rel, person2))
        continue
    elif rel == 'husband' and (p1_gender != 'male' or p2_gender != 'female'):
        invalid_edges_pre_add.append((person1, rel, person2))
        continue
    elif rel == 'wife' and (p1_gender != 'female' or p2_gender != 'male'):
        invalid_edges_pre_add.append((person1, rel, person2))
        continue
    elif rel == 'friend' and p1_gender == 'unknown':
        invalid_edges_pre_add.append((person1, rel, person2))
        continue
    # Factual validation against known relations
    if (person1, rel, person2) in known_relations:
        if not known_relations[(person1, rel, person2)]:
            invalid_edges_pre_add.append((person1, rel, person2))
            continue
    # Check for contradictory relations
    reverse_rel = reverse_relations.get(rel)
    if reverse_rel:
        if G.has_edge(person2, person1):
            existing_rel = G[person2][person1].get('relation')
            if existing_rel != reverse_rel:
                print(f"Warning: Contradictory relation detected: {person1} -> {rel} -> {person2} conflicts with {person2} -> {existing_rel} -> {person1}")
                invalid_edges_pre_add.append((person1, rel, person2))
                continue
    # Add the edge
    if not G.has_edge(person1, person2):
        G.add_edge(person1, person2, relation=rel)
        # Add reverse relation
        if reverse_rel and not G.has_edge(person2, person1) and (person2, reverse_rel, person1) not in invalid_pairs:
            # Validate reverse relation gender constraints
            valid_reverse = True
            if reverse_rel == 'son' and p2_gender != 'male':
                valid_reverse = False
            elif reverse_rel == 'daughter' and p2_gender != 'female':
                valid_reverse = False
            elif reverse_rel == 'father' and p2_gender != 'male':
                valid_reverse = False
            elif reverse_rel == 'mother' and p2_gender != 'female':
                valid_reverse = False
            elif reverse_rel == 'brother' and (p2_gender != 'male' or p1_gender != 'male'):
                valid_reverse = False
            elif reverse_rel == 'sister' and (p2_gender != 'female' or p1_gender != 'female'):
                valid_reverse = False
            elif reverse_rel == 'husband' and (p2_gender != 'male' or p1_gender != 'female'):
                valid_reverse = False
            elif reverse_rel == 'wife' and (p2_gender != 'female' or p1_gender != 'male'):
                valid_reverse = False
            if valid_reverse:
                G.add_edge(person2, person1, relation=reverse_rel)
                reverse_tuple = (person2, reverse_rel, person1)
                split = relation_to_split.get((person1, rel, person2), "train")
                if split == "train":
                    updated_train_relations.add(reverse_tuple)
                elif split == "dev":
                    updated_dev_relations.add(reverse_tuple)
                else:
                    updated_test_relations.add(reverse_tuple)

# Step 3: Post-processing validation
invalid_edges = []
for u, v, data in list(G.edges(data=True)):
    rel = data['relation']
    p1_gender = gender_map.get(u, 'unknown')
    p2_gender = gender_map.get(v, 'unknown')

    # Re-check gender constraints
    if rel == 'son' and p1_gender != 'male':
        invalid_edges.append((u, v, rel))
        continue
    elif rel == 'daughter' and p1_gender != 'female':
        invalid_edges.append((u, v, rel))
        continue
    elif rel == 'father' and p1_gender != 'male':
        invalid_edges.append((u, v, rel))
        continue
    elif rel == 'mother' and p1_gender != 'female':
        invalid_edges.append((u, v, rel))
        continue
    elif rel == 'brother' and (p1_gender != 'male' or p2_gender != 'male'):
        invalid_edges.append((u, v, rel))
        continue
    elif rel == 'sister' and (p1_gender != 'female' or p2_gender != 'female'):
        invalid_edges.append((u, v, rel))
        continue
    elif rel == 'husband' and (p1_gender != 'male' or p2_gender != 'female'):
        invalid_edges.append((u, v, rel))
        continue
    elif rel == 'wife' and (p1_gender != 'female' or p2_gender != 'male'):
        invalid_edges.append((u, v, rel))
        continue

    # Check for invalid entities
    if u in invalid_entities or v in invalid_entities:
        invalid_edges.append((u, v, rel))
        continue

    # Check invalid pairs
    if (u, rel, v) in invalid_pairs:
        invalid_edges.append((u, v, rel))
        continue

    # Factual validation
    if (u, rel, v) in known_relations:
        if not known_relations[(u, rel, v)]:
            invalid_edges.append((u, v, rel))
            continue

# Remove invalid edges
for u, v, rel in invalid_edges:
    if G.has_edge(u, v):
        G.remove_edge(u, v)
        edge_tuple = (u, rel, v)
        updated_train_relations.discard(edge_tuple)
        updated_dev_relations.discard(edge_tuple)
        updated_test_relations.discard(edge_tuple)

# Step 4: Ensure key relations from known_relations are added
for (person1, rel, person2), is_valid in known_relations.items():
    if is_valid and not G.has_edge(person1, person2):
        G.add_edge(person1, person2, relation=rel)
        split = "train"  # Default to train for manually added relations
        updated_train_relations.add((person1, rel, person2))
        # Add reverse relation
        reverse_rel = reverse_relations.get(rel)
        if reverse_rel and not G.has_edge(person2, person1):
            G.add_edge(person2, person1, relation=reverse_rel)
            updated_train_relations.add((person2, reverse_rel, person1))

# Step 5: Rebalance the train/dev/test splits (target: 70/15/15)
all_edges = [(u, v, G[u][v]['relation']) for u, v in G.edges()]
total_edges = len(all_edges)
train_size = int(0.7 * total_edges)
dev_size = int(0.15 * total_edges)
test_size = total_edges - train_size - dev_size

# Shuffle edges to ensure random distribution
import random
random.seed(42)
random.shuffle(all_edges)

# Assign edges to splits
new_train_relations = set()
new_dev_relations = set()
new_test_relations = set()

for i, (u, v, rel) in enumerate(all_edges):
    edge_tuple = (u, rel, v)
    if i < train_size:
        new_train_relations.add(edge_tuple)
    elif i < train_size + dev_size:
        new_dev_relations.add(edge_tuple)
    else:
        new_test_relations.add(edge_tuple)

# Update the relation lists
updated_train_relations = new_train_relations
updated_dev_relations = new_dev_relations
updated_test_relations = new_test_relations

# Log the new split sizes
print("\nUpdated split sizes after rebalancing:")
print(f" - Train: {len(updated_train_relations)} relations")
print(f" - Dev: {len(updated_dev_relations)} relations")
print(f" - Test: {len(updated_test_relations)} relations")

# Step 6: Log graph details for presentation
# Log relation type distribution
relation_types = Counter(data['relation'] for _, _, data in G.edges(data=True))
print("\nRelation type distribution in the graph:")
for rel, count in sorted(relation_types.items()):
    print(f" - {rel}: {count}")

# Verify graph edges for key characters
key_characters = ['krishna', 'arjuna', 'bhima', 'draupadi', 'kunti', 'pandu', 'dhritarashtra', 'subhadra', 'vasudeva', 'karna', 'duryodhana', 'balarama']
print("\nGraph edges for key characters:")
for char in key_characters:
    print(f"\nEdges involving {char}:")
    edges_found = False
    for u, v, data in G.edges(data=True):
        if u == char or v == char:
            edge_str = f"{u} -> {data['relation']} -> {v}"
            print(edge_str)
            edges_found = True
    if not edges_found:
        print("No edges found.")

# Log invalid edges if any
if invalid_edges_pre_add or invalid_edges:
    print("\nInvalid edges removed:")
    for u, v, rel in invalid_edges_pre_add + invalid_edges:
        print(f"{u} -> {rel} -> {v}")
else:
    print("\nNo invalid edges found.")

# Log isolated nodes
isolated_nodes = [node for node in G.nodes if G.degree(node) == 0]
if isolated_nodes:
    print("\nIsolated nodes (no edges):")
    for node in isolated_nodes:
        print(f" - {node}")
else:
    print("\nNo isolated nodes found.")

# Log graph stats
print("\nGraph stats:")
print(f"Nodes: {G.number_of_nodes()}, Edges: {G.number_of_edges()}")

# Step 7: Save the graph and updated relation lists
with open(os.path.join(save_dir, "gita_graph.pkl"), "wb") as f:
    pickle.dump(G, f)
nx.write_gml(G, os.path.join(save_dir, "family_tree.gml"))

# Save updated relation lists
def save_relations_to_file(relations, filename):
    with open(filename, 'w', encoding='utf-8') as f:
        for person1, rel, person2 in relations:
            f.write(f"{person1} -> {rel} -> {person2}\n")

save_relations_to_file(updated_train_relations, os.path.join(save_dir, "train_relations_updated.txt"))
save_relations_to_file(updated_dev_relations, os.path.join(save_dir, "dev_relations_updated.txt"))
save_relations_to_file(updated_test_relations, os.path.join(save_dir, "test_relations_updated.txt"))

# Step 8: Visualize the graph with improved styling
def visualize_graph(G, focus_nodes=None):
    """Visualize the graph with nodes colored by gender and improved edge label styling."""
    net = Network(height="750px", width="100%", notebook=True, directed=True, cdn_resources='remote')
    if focus_nodes:
        subgraph = G.subgraph(focus_nodes)
    else:
        subgraph = G

    # Add nodes with gender-based coloring
    for node in subgraph.nodes:
        gender = gender_map.get(node, "unknown")
        color = "lightblue" if gender == "male" else "pink" if gender == "female" else "gray"
        net.add_node(node, label=node, color=color)

    # Add edges with larger, readable labels
    for u, v, data in subgraph.edges(data=True):
        net.add_edge(u, v, label=data["relation"], color="gray", font={"size": 12})

    visualization_file = os.path.join(save_dir, "family_tree.html")
    net.show(visualization_file)
    return visualization_file

# Visualize the graph for a subset of key characters
focus_nodes = ["arjuna", "krishna", "bhima", "kunti", "pandu", "dhritarashtra", "vasudeva", "karna", "duryodhana"]
visualization_file = visualize_graph(G, focus_nodes)
print(f"\nGraph visualization saved to {visualization_file}")
print(f"Graph saved as {os.path.join(save_dir, 'gita_graph.pkl')} and {os.path.join(save_dir, 'family_tree.gml')}")
print(f"Updated relations saved as {os.path.join(save_dir, 'train_relations_updated.txt')}, {os.path.join(save_dir, 'dev_relations_updated.txt')}, and {os.path.join(save_dir, 'test_relations_updated.txt')}")

print("\nGraph built and visualized. Run the next cell to train the GNN model.")
```

### Train the GNN model

```python
# Cell 4: Train the R-GCN Model
# This cell trains a Relational Graph Convolutional Network (R-GCN) to predict relationships
# between characters in the Mahabharata graph constructed in Cell 3.

import torch
import torch.nn.functional as F
from torch_geometric.data import Data
from torch_geometric.nn import RGCNConv
import numpy as np
from collections import defaultdict, Counter
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import precision_recall_fscore_support, confusion_matrix
import pickle
import networkx as nx
import os

# Device setup
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

# Step 1: Load the graph from Cell 3
save_dir = "/content/drive/My Drive/Gita_Pipeline_Output"
try:
    with open(os.path.join(save_dir, "gita_graph.pkl"), "rb") as f:
        G = pickle.load(f)
except Exception as e:
    raise FileNotFoundError(f"Error loading graph: {e}")

# Load train/dev/test relations (already loaded in Cell 3, assuming they are available)
try:
    train_relations = load_relations_from_file(os.path.join(save_dir, "train_relations_updated.txt"))
    dev_relations = load_relations_from_file(os.path.join(save_dir, "dev_relations_updated.txt"))
    test_relations = load_relations_from_file(os.path.join(save_dir, "test_relations_updated.txt"))
except NameError:
    # If load_relations_from_file is not defined, redefine it here
    def load_relations_from_file(filename):
        relations = []
        try:
            with open(filename, 'r', encoding='utf-8') as f:
                for line in f:
                    line = line.strip()
                    if line:
                        parts = line.split(" -> ")
                        if len(parts) != 3:
                            print(f"Warning: Malformed line in {filename}: {line}")
                            continue
                        person1, rel, person2 = parts
                        person1 = person1.lower()
                        person2 = person2.lower()
                        person1 = name_to_canonical.get(person1, person1)
                        person2 = name_to_canonical.get(person2, person2)
                        relations.append((person1, rel, person2))
        except Exception as e:
            print(f"Error loading relations from {filename}: {e}")
            return []
        return relations

    train_relations = load_relations_from_file(os.path.join(save_dir, "train_relations_cleaned.txt"))
    dev_relations = load_relations_from_file(os.path.join(save_dir, "dev_relations_cleaned.txt"))
    test_relations = load_relations_from_file(os.path.join(save_dir, "test_relations_cleaned.txt"))

# Step 2: Compute mention counts from tagged verses
mention_counts = defaultdict(int)
for split_file in ["train.txt", "dev.txt", "test.txt"]:
    file_path = os.path.join(save_dir, split_file)
    try:
        with open(file_path, 'r', encoding='utf-8') as f:
            for line in f:
                line = line.strip()
                if line:
                    tokens_tags = line.split('\t')
                    if len(tokens_tags) != 2:
                        continue
                    tokens, tags = tokens_tags[0].split(), tokens_tags[1].split()
                    for token, tag in zip(tokens, tags):
                        if tag == 'B-PER':
                            token_lower = token.lower()
                            canonical_name = name_to_canonical.get(token_lower, token_lower)
                            if canonical_name in person_names_set:
                                mention_counts[canonical_name] += 1
    except Exception as e:
        print(f"Error processing {file_path}: {e}")

# Step 3: Prepare data for R-GCN
node_list = list(G.nodes())
node2idx = {node: idx for idx, node in enumerate(node_list)}
num_nodes = len(node_list)

# Node features with learned embeddings
node_features = []
for node in node_list:
    degree = G.degree(node)
    mention_count = mention_counts[node]
    in_degree = G.in_degree(node)
    out_degree = G.out_degree(node)
    clustering = nx.clustering(G.to_undirected(), node)
    gender = 1.0 if gender_map.get(node, 'unknown') == 'male' else 0.0 if gender_map.get(node, 'unknown') == 'female' else 0.5
    node_features.append([degree, mention_count, in_degree, out_degree, clustering, gender])

# Normalize features
node_features = np.array(node_features)
node_features[:, :-1] = (node_features[:, :-1] - node_features[:, :-1].mean(axis=0)) / (node_features[:, :-1].std(axis=0) + 1e-8)
x = torch.tensor(node_features, dtype=torch.float).to(device)

# Learned node embeddings
embed_dim = 16  # Reduced embedding dimension for smaller graph
char_embedding = torch.nn.Embedding(num_nodes, embed_dim).to(device)
char_ids = torch.arange(num_nodes).to(device)
char_embed = char_embedding(char_ids)
x = torch.cat([x, char_embed], dim=1)  # Append 16-dim embeddings

# Prepare edges and edge labels
edges = []
edge_labels = []
relation_types = list(set([data["relation"] for _, _, data in G.edges(data=True)]))
relation_encoder = LabelEncoder()
relation_encoder.fit(relation_types)

for u, v, data in G.edges(data=True):
    edges.append([node2idx[u], node2idx[v]])
    edge_labels.append(relation_encoder.transform([data["relation"]])[0])

edges = torch.tensor(edges, dtype=torch.long).t().contiguous().to(device)
edge_labels = torch.tensor(edge_labels, dtype=torch.long).to(device)

# Class weights for handling imbalanced classes
unique_labels, label_counts = np.unique(edge_labels.cpu().numpy(), return_counts=True)
class_weights = 1.0 / (label_counts + 1e-8) ** 0.5
class_weights = class_weights / class_weights.sum() * len(unique_labels)
class_weights = torch.tensor(class_weights, dtype=torch.float).to(device)

# Log class weights
print("Class weights for relations:")
for rel, weight in zip(relation_encoder.classes_, class_weights.cpu().numpy()):
    print(f" - {rel}: {weight:.4f}")

# Create PyTorch Geometric Data object
data = Data(x=x, edge_index=edges, edge_attr=edge_labels, num_nodes=num_nodes).to(device)

# Train/dev/test masks
edge_indices = torch.arange(edges.shape[1])
train_mask = torch.zeros(edges.shape[1], dtype=torch.bool).to(device)
dev_mask = torch.zeros(edges.shape[1], dtype=torch.bool).to(device)
test_mask = torch.zeros(edges.shape[1], dtype=torch.bool).to(device)

for i, (u, v) in enumerate(edges.t()):
    u_node, v_node = node_list[u], node_list[v]
    relation = G.get_edge_data(u_node, v_node)["relation"]
    rel_tuple = (u_node, relation, v_node)
    if rel_tuple in train_relations:
        train_mask[i] = 1
    elif rel_tuple in dev_relations:
        dev_mask[i] = 1
    elif rel_tuple in test_relations:
        test_mask[i] = 1
    else:
        print(f"Warning: Edge {rel_tuple} not found in any split, assigning to train")
        train_mask[i] = 1

data.train_mask = train_mask
data.dev_mask = dev_mask
data.test_mask = test_mask

# Log dataset sizes
print(f"\nGraph size: {num_nodes} nodes, {edges.shape[1]} edges")
print(f"Number of relation types: {len(relation_types)}")
print(f"Train edges: {data.train_mask.sum().item()}, Dev edges: {data.dev_mask.sum().item()}, Test edges: {data.test_mask.sum().item()}")

# Step 4: Define R-GCN Model
class RGCN(torch.nn.Module):
    def __init__(self, in_channels, hidden_channels, out_channels, num_relations, num_classes):
        super(RGCN, self).__init__()
        self.conv1 = RGCNConv(in_channels, hidden_channels, num_relations)
        self.ln1 = torch.nn.LayerNorm(hidden_channels)
        self.conv2 = RGCNConv(hidden_channels, out_channels, num_relations)
        self.ln2 = torch.nn.LayerNorm(out_channels)
        self.dropout = torch.nn.Dropout(0.3)  # Reduced dropout for better learning
        self.fc = torch.nn.Linear(out_channels * 2, num_classes)

    def forward(self, data):
        x, edge_index, edge_type = data.x, data.edge_index, data.edge_attr
        x = self.conv1(x, edge_index, edge_type)
        x = self.ln1(x)
        x = F.relu(x)
        x = self.dropout(x)
        x = self.conv2(x, edge_index, edge_type)
        x = self.ln2(x)
        x = F.relu(x)
        x = self.dropout(x)
        edge_embeddings = torch.cat([x[edge_index[0]], x[edge_index[1]]], dim=1)
        out = self.fc(edge_embeddings)
        return out

# Initialize model
num_classes = len(relation_types)
num_relations = len(relation_types)
in_channels = x.shape[1]  # 6 features + 16-dim embedding
model = RGCN(in_channels=in_channels, hidden_channels=128, out_channels=64, num_relations=num_relations, num_classes=num_classes).to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=0.001, weight_decay=5e-4)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=300, eta_min=1e-6)

# Label smoothing loss
class LabelSmoothingLoss(torch.nn.Module):
    def __init__(self, smoothing=0.1):
        super(LabelSmoothingLoss, self).__init__()
        self.smoothing = smoothing
        self.cls = num_classes

    def forward(self, pred, target):
        pred = pred.log_softmax(dim=-1)
        with torch.no_grad():
            true_dist = torch.zeros_like(pred)
            true_dist.fill_(self.smoothing / (self.cls - 1))
            true_dist.scatter_(1, target.unsqueeze(1), 1.0 - self.smoothing)
        return torch.mean(torch.sum(-true_dist * pred, dim=-1))

criterion = LabelSmoothingLoss(smoothing=0.1)

# Training and evaluation functions
def train():
    model.train()
    optimizer.zero_grad()
    out = model(data)
    loss = criterion(out[data.train_mask], data.edge_attr[data.train_mask])
    loss.backward(retain_graph=True)
    optimizer.step()
    scheduler.step()
    return loss.item()

def evaluate(mask, log_misclassifications=False, epoch=None, split_name=""):
    model.eval()
    misclassifications = []
    with torch.no_grad():
        out = model(data)
        pred = out.argmax(dim=1)
        correct = (pred[mask] == data.edge_attr[mask]).sum()
        acc = int(correct) / int(mask.sum()) if mask.sum() > 0 else 0
        # Compute precision, recall, F1 for each class
        true_labels = data.edge_attr[mask].cpu().numpy()
        pred_labels = pred[mask].cpu().numpy()
        precision, recall, f1, _ = precision_recall_fscore_support(true_labels, pred_labels, average=None, labels=np.arange(num_classes), zero_division=0)
        # Compute loss
        loss = criterion(out[mask], data.edge_attr[mask]).item()
        # Log misclassifications
        if log_misclassifications:
            incorrect = (pred[mask] != data.edge_attr[mask]).nonzero(as_tuple=True)[0]
            for idx in incorrect:
                edge_idx = torch.where(mask)[0][idx]
                u, v = edges[:, edge_idx]
                true_rel = relation_encoder.inverse_transform([data.edge_attr[edge_idx].item()])[0]
                pred_rel = relation_encoder.inverse_transform([pred[edge_idx].item()])[0]
                misclass = f"{node_list[u]} -> {true_rel} -> {node_list[v]} (Predicted: {pred_rel})"
                misclassifications.append(misclass)
    return acc, loss, precision, recall, f1, true_labels, pred_labels, misclassifications

# Training loop
best_dev_acc = 0
best_dev_f1 = 0
best_epoch = 0
patience = 100
wait = 0
misclassification_log = []

print("\nStarting training...")
for epoch in range(500):
    loss = train()
    train_acc, train_loss, train_precision, train_recall, train_f1, _, _, _ = evaluate(data.train_mask)
    dev_acc, dev_loss, dev_precision, dev_recall, dev_f1, dev_true, dev_pred, dev_misclass = evaluate(data.dev_mask, log_misclassifications=True, epoch=epoch, split_name="Dev")
    test_acc, test_loss, test_precision, test_recall, test_f1, _, _, _ = evaluate(data.test_mask)
    lr = scheduler.get_last_lr()[0]
    dev_f1_macro = np.mean(dev_f1)

    if epoch % 10 == 0:
        print(f"Epoch {epoch}, Loss: {loss:.4f}, LR: {lr:.6f}")
        print(f"Train - Acc: {train_acc:.4f}, Loss: {train_loss:.4f}")
        print(f"Dev - Acc: {dev_acc:.4f}, Loss: {dev_loss:.4f}, Macro F1: {dev_f1_macro:.4f}")
        print(f"Test - Acc: {test_acc:.4f}, Loss: {test_loss:.4f}")
        if dev_misclass:
            print("Dev Misclassifications (up to 5):")
            for misclass in dev_misclass[:5]:
                print(misclass)
        misclassification_log.extend([(epoch, misclass) for misclass in dev_misclass])

    if dev_acc > best_dev_acc:
        best_dev_acc = dev_acc
        best_dev_f1 = dev_f1_macro
        best_epoch = epoch
        wait = 0
        # Save the best model
        torch.save({
            'model_state_dict': model.state_dict(),
            'node2idx': node2idx,
            'node_list': node_list,
            'num_nodes': num_nodes,
            'num_classes': num_classes,
            'node_features': x
        }, os.path.join(save_dir, 'gita_r_gcn_model.pt'))  # Updated filename to 'r_gcn'
        # Save best metrics
        best_metrics = {
            'epoch': best_epoch,
            'dev_acc': dev_acc,
            'dev_f1': dev_f1,
            'dev_precision': dev_precision,
            'dev_recall': dev_recall,
            'test_acc': test_acc,
            'test_f1': test_f1,
            'test_precision': test_precision,
            'test_recall': test_recall
        }
    else:
        wait += 1
        if wait >= patience:
            print(f"Early stopping at epoch {epoch}. Best dev accuracy: {best_dev_acc:.4f}, Best dev F1: {best_dev_f1:.4f} at epoch {best_epoch}")
            break

# Step 5: Final evaluation on test set with detailed metrics
print("\nFinal evaluation on test set:")
test_acc, test_loss, test_precision, test_recall, test_f1, test_true, test_pred, test_misclass = evaluate(data.test_mask, log_misclassifications=True, split_name="Test")
test_f1_macro = np.mean(test_f1)
print(f"Test Accuracy: {test_acc:.4f}, Test Loss: {test_loss:.4f}, Test Macro F1: {test_f1_macro:.4f}")

# Log per-class metrics
print("\nPer-class metrics on test set:")
for i, rel in enumerate(relation_encoder.classes_):
    print(f"{rel} - Precision: {test_precision[i]:.4f}, Recall: {test_recall[i]:.4f}, F1: {test_f1[i]:.4f}")

# Compute and log confusion matrix
conf_matrix = confusion_matrix(test_true, test_pred, labels=np.arange(num_classes))
print("\nConfusion Matrix (Test Set):")
print(conf_matrix)

# Analyze common misclassification patterns
misclass_pairs = Counter()
for misclass in test_misclass:
    true_rel = misclass.split(" -> ")[1]
    pred_rel = misclass.split("(Predicted: ")[1].rstrip(")")
    misclass_pairs[(true_rel, pred_rel)] += 1
print("\nMost common misclassification patterns (Test Set):")
for (true_rel, pred_rel), count in misclass_pairs.most_common(5):
    print(f"True: {true_rel}, Predicted: {pred_rel}, Count: {count}")

# Save misclassifications
with open(os.path.join(save_dir, 'misclassifications.txt'), 'w') as f:
    f.write("Dev Misclassifications:\n")
    for epoch, misclass in misclassification_log:
        f.write(f"Epoch {epoch}: {misclass}\n")
    f.write("\nTest Misclassifications:\n")
    for misclass in test_misclass:
        f.write(f"{misclass}\n")

# Save model, encoder, and metrics
with open(os.path.join(save_dir, 'relation_encoder_classes.pkl'), 'wb') as f:
    pickle.dump(relation_encoder.classes_, f)
with open(os.path.join(save_dir, 'best_metrics.pkl'), 'wb') as f:
    pickle.dump(best_metrics, f)

print(f"\nModel saved as {os.path.join(save_dir, 'gita_r_gcn_model.pt')}")
print(f"Relation encoder saved as {os.path.join(save_dir, 'relation_encoder_classes.pkl')}")
print(f"Misclassifications saved as {os.path.join(save_dir, 'misclassifications.txt')}")
print(f"Best metrics saved as {os.path.join(save_dir, 'best_metrics.pkl')}")

print("\nR-GCN model trained. Run the next cell to predict relationships.")
```

```python
# Cell 5: Predict Relationships

# Step 19: Test the Model with Relationship Inference
def predict_relationship(model, person1, person2, node2idx, relation_encoder, G, reverse=False):
    person1 = name_to_canonical.get(person1.lower(), person1.lower())
    person2 = name_to_canonical.get(person2.lower(), person2.lower())

    if person1 not in node2idx or person2 not in node2idx:
        return f"No relationship found between {person1} and {person2}", 0.0

    reverse_relations = {
        "father": "son", "mother": "daughter", "son": "father", "daughter": "mother",
        "grandfather": "grandson", "grandson": "grandfather", "brother": "brother",
        "sister": "sister", "cousin": "cousin", "brother-in-law": "brother-in-law",
        "sister-in-law": "sister-in-law", "husband": "wife", "wife": "husband",
        "maternal uncle": "nephew/niece", "father-in-law": "son-in-law", "family": "family"
    }

    if not reverse and G.has_edge(person1, person2):
        relation = G.get_edge_data(person1, person2)["relation"]
        return f"{person1} is the {relation} of {person2}", 1.0
    elif reverse and G.has_edge(person2, person1):
        relation = G.get_edge_data(person2, person1)["relation"]
        reverse_relation = reverse_relations.get(relation, "unknown relation")
        return f"{person1} is the {reverse_relation} of {person2}", 1.0

    idx1, idx2 = node2idx[person1], node2idx[person2]
    inference_edge = torch.tensor([[idx1], [idx2]], dtype=torch.long).to(device)
    inference_data = Data(x=data.x, edge_index=inference_edge).to(device)

    model.eval()
    with torch.no_grad():
        out = model(inference_data)
        probs = F.softmax(out, dim=1)
        pred = out.argmax(dim=1)
        confidence = probs[0, pred[0]].item()
        relation = relation_encoder.inverse_transform(pred.cpu().numpy())[0]
    if reverse:
        relation = reverse_relations.get(relation, "unknown relation")
    return f"Predicted: {person1} is the {relation} of {person2}", confidence

def determine_relationships(user_input, model, node2idx, relation_encoder, G):
    names = [name.strip().lower() for name in user_input.replace(" ", ",").split(",") if name.strip()]
    if len(names) < 2:
        return [("Please provide at least two names (e.g., arjuna, krishna).", 0.0)]

    first_person = names[0]
    other_persons = names[1:]
    results = []
    for person in other_persons:
        relationship, confidence = predict_relationship(model, first_person, person, node2idx, relation_encoder, G, reverse=False)
        if "No relationship found" in relationship or "Predicted" in relationship:
            relationship, confidence = predict_relationship(model, first_person, person, node2idx, relation_encoder, G, reverse=True)
        results.append((relationship, confidence))
    return results

# Step 20: Test Specific Relationships
test_pairs = [
    ("krishna", "bhima"),  # Should be cousin
    ("arjuna", "draupadi"),  # Should be husband
    ("kunti", "arjuna"),  # Should be mother
    ("dhritarashtra", "pandu"),  # Should be brother
    ("subhadra", "krishna")  # Should be sister
]

print("Testing key relationships:")
for person1, person2 in test_pairs:
    relationship, confidence = predict_relationship(model, person1, person2, node2idx, relation_encoder, G, reverse=False)
    print(f"{relationship} (Confidence: {confidence:.2f})")
    if "Predicted" in relationship:
        relationship, confidence = predict_relationship(model, person1, person2, node2idx, relation_encoder, G, reverse=True)
        print(f"Reverse: {relationship} (Confidence: {confidence:.2f})")

# Interactive testing
print("\nEnter names separated by commas (e.g., arjuna, krishna, draupadi):")
user_input = input()
relationships = determine_relationships(user_input, model, node2idx, relation_encoder, G)
for rel, conf in relationships:
    print(f"{rel} (Confidence: {conf:.2f})")

print(f"All output files are saved in {save_dir}. You can access them from your Google Drive.")
```
